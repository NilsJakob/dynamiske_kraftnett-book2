# Fra idé til ELK330

I prosessen med å lage emne ELK330 ble jeg inspirert av Porfessor Mattia Marinelli ved Dansk Teknisk Universitet (DTU), som har laget et tilsvarende emne på DTU.

Han hadde to ledesnorerunder utviklingen av emnet:

1. Hva slags kurs ville jeg selv ønsket å ta som student

2. Hva kan jeg bidra med fra min egen bakgrunn/ekspertise inn i faget

Punkt 1 har jeg alltid som student ønsket meg verktøy i fra utdanningen jeg kan ta med meg videre.

- Git
- Power Factory

Punkt 2 vil jeg bringe inn min bakgrunn med applikert maskinlæring og dataanalyse. En av grunnene til at jeg valgte å studere maskinlæring i mine master-studier var at jeg ønsket et verktøy.

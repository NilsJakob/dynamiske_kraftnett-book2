## Oppgave 1: Den generelle modellen

Vi antar at belastningen kan beskrives som

$$ L(t) = L_0 + \sum_{i=1}^{n} A_i \exp\!\left(-\frac{(t-\mu_i)^2}{2\sigma_i^2}\right).
$$

Der:

- $L_0$ er grunnlasten,
- $A_i$ er amplituden til komponent $i$,
- $\mu_i$ er tidspunktet for topp,
- $\sigma_i$ er bredden på komponenten.

### Oppgaver

1. Forklar med egne ord hvordan parameterne $A_i$, $\mu_i$ og $\sigma_i$ påvirker kurven.
2. Implementer modellen i Python.
3. Eksperimenter med parameterverdiene og dokumenter observasjonene.

## Oppgave 7: Bias-varians-avveining

Vi antar at det finnes en ukjent sann sammenheng

$$
y=f(x)+\varepsilon,
$$

og at modellen vår er

$$
\hat f(x).
$$

Diskuter:

1. Hva representerer $f(x)$?
2. Hva representerer $\hat f(x)$
3. Hva skjer dersom modellen har for få komponenter?
4. Hva skjer dersom modellen har svært mange komponenter?
5. Hvor mener du at din modell befinner seg på skalaen mellom høy bias og høy varians?

Avslutt med å kommentere utsagnet:

> En modell med høy bias lærer for lite av dataene. En modell med høy varians lærer for mye av dataene. En god modell lærer akkurat nok.

# Øving 2: Branch, Push, Pull og Merge med Git og GitHub

## Læringsmål

Etter denne øvingen skal du kunne:

- Opprette og bytte mellom brancher
- Gjøre endringer og opprette commits
- Push endringer til GitHub
- Pull endringer fra GitHub
- Merge brancher
- Forstå hvordan Git brukes i samarbeid

---

## Del 1: Opprette en branch og merge den

### 1. Klon repositoryet

```bash
git clone <repo-url>
cd <repo-navn>
```

### 2. Opprett en ny branch

```bash
git checkout -b feature-1
```

### 3. Gjør en endring

Legg til en linje i en eksisterende fil eller opprett en ny fil.

### 4. Lagre endringen som en commit

```bash
git add .
git commit -m "La til feature 1"
```

### 5. Bytt tilbake til main

```bash
git checkout main
```

### 6. Merge branchen

```bash
git merge feature-1
```

### 7. Se commit-historikken

```bash
git log --oneline --graph
```

### Oppgave

Ta et skjermbilde av resultatet fra:

```bash
git log --oneline --graph
```

---

## Del 2: Push og Pull

### 1. Push endringer til GitHub

```bash
git push origin main
```

### 2. Opprett en ny branch

```bash
git checkout -b feature-2
```

### 3. Gjør en ny endring

Rediger en fil i repositoryet.

### 4. Commit endringen

```bash
git add .
git commit -m "La til feature 2"
```

### 5. Push branchen til GitHub

```bash
git push origin feature-2
```

### 6. Bytt tilbake til main

```bash
git checkout main
```

### 7. Hent siste versjon fra GitHub

```bash
git pull origin main
```

## Del 3: Simuler to utviklere

Denne delen gjennomføres i grupper på to studenter.

Begge studentene kloner samme repository.

### Student A

Legg til en linje i filen `notes.txt`:

```text
Lagt til av Student A
```

Deretter:

```bash
git add .
git commit -m "Oppdatering fra Student A"
git push
```

### Student B

Legg til en annen linje i samme fil:

```text
Lagt til av Student B
```

Deretter:

```bash
git add .
git commit -m "Oppdatering fra Student B"
```

Prøv å hente siste versjon:

```bash
git pull
```

Dersom Git rapporterer en konflikt, må konflikten løses før arbeidet kan fortsette.

Når konflikten er løst:

```bash
git add .
git commit
git push
```

## Del 4: Pull Request på GitHub

### 1. Opprett en ny branch

```bash
git checkout -b min-feature
```

### 2. Gjør en endring og commit den

```bash
git add .
git commit -m "Lagde min feature"
```

### 3. Push branchen

```bash
git push origin min-feature
```

### 4. Opprett en Pull Request

Gå til GitHub og opprett en Pull Request fra `min-feature` til `main`.

### 5. Merge Pull Requesten

Gjennomfør merge på GitHub.

### 6. Oppdater ditt lokale repository

```bash
git checkout main
git pull
```

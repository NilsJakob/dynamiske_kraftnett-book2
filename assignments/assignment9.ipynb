# Øving 9

En generator som opererer ved 50 Hz leverer en mekanisk effekt på

$$
P_m = 0.9 \text{ p.u.}
$$

til en uendelig sterk skinne gjennom et nett der resistansen kan neglisjeres.

Før feilen er den maksimalt overførbare effekten

$$
P_{\max I}=1.6 \text{ p.u.}
$$

Under feilen reduseres den maksimalt overførbare effekten til

$$
P_{\max II}=0.3 \text{ p.u.}
$$

Etter at feilen er koblet ut, blir den maksimalt overførbare effekten

$$
P_{\max III}=1.2 \text{ p.u.}
$$

Generatorens treghetskonstant er

$$
H=3.0 \text{ p.u.}
$$

Feilen kobles ut etter

$$
t_c=0.15\text{ s}
$$

Bruk:

1. Likearealkriteriet til å bestemme den kritiske utkoblingsvinkelen.
2. Punkt-til-punkt-metoden (step-by-step solution) til å beregne svingskurven.

Bruk tidssteg

$$
\Delta t=0.05\text{ s}
$$

---

# Løsningsgrunnlag

## Likearealkriteriet

Startvinkelen er

$$
\delta_0=
\sin^{-1}
\left(
\frac{P_m}{P_{\max I}}
\right)
$$

Maksimal rotorvinkel ved kritisk stabilitet er

$$
\delta_{\max}
=
180^\circ
-
\sin^{-1}
\left(
\frac{P_m}{P_{\max III}}
\right)
$$

Den kritiske utkoblingsvinkelen finnes fra

$$
\cos \delta_{cr}
=
\frac{
P_m(\delta_{\max}-\delta_0)
-
P_{\max II}\cos\delta_0
+
P_{\max III}\cos\delta_{\max}
}
{
P_{\max III}-P_{\max II}
}
$$

---

## Svingskurve

Bruk

$$
M
=
\frac{H}{180f}
$$

med

$$
f=50\text{ Hz}
$$

og beregn

$$
\frac{(\Delta t)^2}{M}
$$

Den akselererende effekten under feilen er

$$
\Delta P
=
P_m
-
P_{\max II}\sin\delta
$$

Etter feilutkobling er

$$
\Delta P
=
P_m
-
P_{\max III}\sin\delta
$$

Rotorvinkelen beregnes trinnvis fra

$$
\Delta\delta_n
=
\Delta\delta_{n-1}
+
\frac{(\Delta t)^2}{M}
\Delta P_{n-1}
$$

og

$$
\delta_n
=
\delta_{n-1}
+
\Delta\delta_n
$$

---

# Svar

Kontroller at du finner omtrent

$$
\delta_0 \approx 34.2^\circ
$$

og

$$
\delta_{\max} \approx 131.4^\circ
$$

Før du fortsetter beregningen av

$$
\delta_{cr}
$$

og svingskurven.

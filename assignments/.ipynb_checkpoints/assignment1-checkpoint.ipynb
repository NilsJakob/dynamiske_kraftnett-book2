## Øving 1: Opprett GitHub-repo

### Læringsmål

Etter denne øvingen skal du kunne:

- opprette en GitHub-konto,
- opprette et GitHub-repo,
- dokumentere arbeidet ved hjelp av LaTeX,
- generere en PDF-rapport.

---

### Oppgave 1: Opprett en GitHub-bruker

Dersom du ikke allerede har en GitHub-konto:

1. Gå til:

   https://github.com

2. Klikk **Sign up**.

3. Opprett en brukerkonto.

4. Verifiser kontoen og logg inn.

---

### Oppgave 2: Opprett repo

Opprett et nytt repo på GitHub.

Repoet skal hete:

```text
ELK330
```

eller

```text
elk330
```

Repo skal være:

- Public
- uten README-fil

Når repoet er opprettet, ta et skjermbilde som viser:

- repo-navn
- GitHub-kontoen

---

### Oppgave 3: Dokumenter arbeidet i LaTeX

Opprett filen:

```text
oving1.tex
```

og inkluder skjermbildet ved hjelp av:

```latex
\includegraphics{}
```



### Levere

Last opp følgende fil i canvas:

```text
oving1.pdf
```

## Øving 2: Klon Git-repo

### Læringsmål

Etter denne øvingen skal du kunne:

- klone et Git-repositorium fra GitHub,
- navigere i mappestrukturen,
- åpne datafiler i et repo,
- hente oppdateringer med Git.

---

### Bakgrunn

I dette emnet distribueres datasett gjennom GitHub.

I stedet for å laste ned filer manuelt skal vi bruke Git til å hente og oppdatere data.

---

### Oppgave 1: Opprett en arbeidsmappe

Åpne Git Bash og gå til katalogen der du ønsker å lagre repoet.

Eksempel:

```bash
cd /c/repos
```

---

### Oppgave 2: Klon repo

Klon datasett-repoet:

```bash
git clone https://github.com/NilsJakob/power-system-data.git
```

Git oppretter nå en lokal kopi av repoet.

---

### Oppgave 3: Gå inn i repo

```bash
cd power-system-data
```

Vis innholdet:

```bash
ls
```

Du bør se mapper som:

```text
docs/
images/
load/
src/
```

---

### Oppgave 4: Undersøk datasettet

Vis innholdet i mappen `load`:

```bash
ls load
```

Åpne datasettet i et valgfritt program og studer innholdet:

```text
forbruk_2025.csv
```

Beskriv:

- hvor mange kolonner datasettet inneholder,
- hvilke størrelser som er registrert,
- tidsoppløsningen i datasettet.

---

### Oppgave 5: Åpne repositoryet i VS Code

Start Visual Studio Code:

```bash
code .
```

Undersøk:

- mappestrukturen,
- README-filen,
- Python-skriptet i `src`.

Beskriv kort hva skriptet brukes til.

---

### Oppgave 6: Hent oppdateringer

Git brukes ofte til å hente nye versjoner av data.

Kjør:

```bash
git pull
```

Forklar med egne ord hva kommandoen gjør.

---

### Refleksjon

Diskuter følgende spørsmål:

1. Hva er fordelene med å distribuere data gjennom GitHub?
2. Hvorfor kan Git være bedre enn å sende datasett som e-postvedlegg?
3. Hva er forskjellen mellom:

```bash
git clone
```

og

```bash
git pull
```

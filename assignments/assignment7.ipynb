# Øving 7 – Beregning av elektromekaniske konstanter

## Oppgave

Betrakt en 2-polet, 50 Hz, 13,8 kV turbogenerator. Generatoren har merkedata på **60 MVA** og **0,85 etterslepende effektfaktor (lagging)**.

Den synkrone maskinen har et treghetsmoment på

$$
I = 4200\; 	ext{kg}\cdot	ext{m}^2
$$

### (i)
Beregn den mekaniske rotasjonshastigheten $$(\omega_{mec,rpm})$$.

### (ii)
Beregn den synkrone vinkelhastigheten $$(\omega_s)$$ i rad/s.

### (iii)
Beregn den kinetiske energien (KE) lagret ved synkron hastighet (MJ).

### (iv)
Beregn treghetskonstanten $$H$$ i MJ/MVA.

### (v)
Beregn treghetskonstanten $$H$$ i MJ/MW.

### (vi)
Beregn momentumkonstanten $$M$$ i MJ·s per elektrisk grad.

---

## Fasit

### (i)

$$
\omega_{mec} = rac{120f}{P} = rac{120\cdot 50}{2} = 3000\;	ext{rpm}
$$

**Svar:**

$$
\omega_{mec} = 3000.0\;	ext{rpm}
$$

### (ii)

$$
\omega_s = 2\pi f = 2\pi \cdot 50 = 314.1593\;	ext{rad/s}
$$

### (iii)

$$
KE = rac{1}{2}I\omega_s^2
$$

$$
KE = 207.2617\;	ext{MJ}
$$

### (iv)

$$
H = rac{KE}{S_{nom}}
$$

$$
H = rac{207.2617}{60} = 3.4544\;	ext{MJ/MVA}
$$

### (v)

$$
P_{nom} = S_{nom}\cosarphi = 60\cdot0.85 = 51.00\;	ext{MW}
$$

$$
H = rac{KE}{P_{nom}} = 4.0640\;	ext{MJ/MW}
$$

### (vi)

$$
M = rac{S_{nom}H}{180f}
$$

$$
M = 0.0230\;	ext{MJ}\cdot	ext{s/el.grad}
$$

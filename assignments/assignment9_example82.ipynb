# Eksempel 8.2

En generator som opererer ved 50 Hz leverer en effekt på $1$ p.u. til en uendelig sterk skinne (infinite bus) gjennom et nett der resistansen kan neglisjeres.

Det oppstår en feil som reduserer den maksimalt overførbare effekten til $0.4$ p.u. Under normal drift før feilen var den maksimalt overførbare effekten $1.8$ p.u., og etter at feilen er koblet ut er den $1.3$ p.u.

Ved bruk av **likearealkriteriet (Equal Area Criterion)** skal den kritiske utkoblingsvinkelen bestemmes.

## Løsning

Maksimal overførbar effekt i de tre driftstilstandene er:

- Før feil: $P_{\max I}=1.8$ p.u.
- Under feil: $P_{\max II}=0.4$ p.u.
- Etter feilutkobling: $P_{\max III}=1.3$ p.u.

Den mekaniske effekten antas konstant:

$$
P_m = 1.0 \text{ p.u.}
$$

### Startvinkel før feilen

Rotorvinkelen før feilen er

$$
\delta_0
=
\sin^{-1}
\left(
\frac{P_m}{P_{\max I}}
\right)
$$

$$
\delta_0
=
\sin^{-1}
\left(
\frac{1.0}{1.8}
\right)
=
33.8^\circ
$$

### Maksimal rotorvinkel ved kritisk stabilitet

Ved kritisk stabilitet blir maksimal rotorvinkel

$$
\delta_{\max}
=
180^\circ
-
\sin^{-1}
\left(
\frac{P_m}{P_{\max III}}
\right)
$$

$$
\delta_{\max}
=
180^\circ
-
\sin^{-1}
\left(
\frac{1.0}{1.3}
\right)
=
129.72^\circ
$$

### Kritisk utkoblingsvinkel

Stevensons ligning gir

$$
\cos \delta_{cr}
=
\frac{
P_m(\delta_{\max}-\delta_0)
-
P_{\max II}\cos\delta_0
+
P_{\max III}\cos\delta_{\max}
}
{
P_{\max III}-P_{\max II}
}
$$

der $\delta_0$ og $\delta_{\max}$ må uttrykkes i radianer.

Innsetting av tallverdier gir

$$
\cos \delta_{cr}
=
\frac{
1(2.26-0.59)
-
0.4\cdot 0.831
+
1.3\cdot(-0.64)
}
{1.3-0.4}
$$

$$
\cos \delta_{cr}
=
0.562
$$

Dermed blir den kritiske utkoblingsvinkelen

$$
\delta_{cr}
=
\cos^{-1}(0.562)
=
55.8^\circ
$$

## Resultat

$$
\boxed{\delta_{cr}=55.8^\circ}
$$

Generatoren må derfor koble ut feilen før rotorvinkelen når omtrent $55.8^\circ$ for å opprettholde transient stabilitet.

# Eksempel 8.3

For systemet beskrevet i Eksempel 8.2 er treghetskonstanten for generator og turbin

$$
H=2.7\ \text{p.u.}
$$

Bestem svingskurven (swing curve) når feilutkoblingstiden er

$$
t_c = 125\ \text{ms}
$$

## Løsning

Gitt:

$$
H=2.7\ \text{p.u.}
$$

$$
f=50\ \text{Hz}
$$

$$
G=1\ \text{p.u.}
$$

Den normaliserte treghetskonstanten blir

$$
M=\frac{HG}{180f}
=\frac{2.7\times 1}{180\times 50}
=3\times10^{-4}
$$

Vi velger tidssteget

$$
\Delta t = 0.05\ \text{s}
$$

Dermed fås

$$
\frac{(\Delta t)^2}{M}
=
\frac{(0.05)^2}{3\times10^{-4}}
=
8.33
$$

Startvinkelen er fra Eksempel 8.2

$$
\delta_0
=
\sin^{-1}
\left(
\frac{1}{1.8}
\right)
=
33.8^\circ
$$

Like før feilen er den akselererende effekten

$$
\Delta P = 0
$$

Umiddelbart etter feilen blir

$$
\Delta P
=
P_m-P_{\max II}\sin\delta_0
$$

$$
\Delta P
=
1-0.4\sin 33.8^\circ
=
0.78\ \text{p.u.}
$$

For første tidsintervall benyttes halv verdi

$$
\Delta P=\frac{0.78}{2}
=
0.39\ \text{p.u.}
$$

Den trinnvise løsningen beregnes fra

$$
\Delta\delta_n
=
\Delta\delta_{n-1}
+
\frac{(\Delta t)^2}{M}
\Delta P_{n-1}
$$

og

$$
\delta_n
=
\delta_{n-1}
+
\Delta\delta_n
$$

---

## Intervall 1

$$
t=0.05\ \text{s}
$$

$$
\Delta P=0.39
$$

$$
\Delta\delta_{0}
=
0+8.33(0.39)
=
3.25^\circ
$$

$$
\delta_{0.05}
=
33.8+3.25
=
37.05^\circ
$$

---

## Intervall 2

$$
t=0.10\ \text{s}
$$

$$
\Delta P
=
1-0.4\sin 37.05^\circ
=
0.76
$$

$$
\Delta\delta_{0.10}
=
3.25+8.33(0.76)
=
9.58^\circ
$$

$$
\delta_{0.10}
=
37.05+9.58
=
46.63^\circ
$$

---

## Intervall 3

$$
t=0.15\ \text{s}
$$

$$
\Delta P
=
1-0.4\sin 46.63^\circ
=
0.71
$$

$$
\Delta\delta_{0.15}
=
9.58+8.33(0.71)
=
15.49^\circ
$$

$$
\delta_{0.15}
=
46.63+15.49
=
62.12^\circ
$$

Feilen kobles ut etter

$$
t_c=0.125\ \text{s}
$$

Siden dette tidspunktet ligger midt i intervallet $0.10$ s til $0.15$ s, oppstår et brudd i effekt-vinkel-kurven som krever særskilt behandling.

Etter feilutkobling benyttes post-fault-kurven

$$
P_e=P_{\max III}\sin\delta
$$

slik at

$$
\Delta P
=
P_m-P_{\max III}\sin\delta
=
1-1.3\sin\delta
$$

---

## Intervall 4

$$

$$

=
1-1.3\sin 62.12^\circ
=
-0.149
$$

=
15.49+8.33(-0.149)

14.25^\circ
$$

$$
\delta_{0.20}
=
62.12+14.25
=
76.37^\circ
$$

---

## Intervall 5

t=0.25\ \text{s}
$$

$$
\Delta P
=
1-1.3\sin 76.37^\circ
=
-0.26
$$

$$
\Delta\delta_{0.25}
=
14.25+8.33(-0.26)
=
12.08^\circ
$$

$$
\delta_{0.25}
=
76.37+12.08
=
88.39^\circ
$$

---

## Intervall 6

$$
t=0.30\ \text{s}
$$

$$

=
1-1.3\sin 88.39^\circ
=
-0.30
$$

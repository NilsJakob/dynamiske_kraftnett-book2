# Hva er versjonskontroll?

**Versjonskontroll** er et system for å holde orden på endringer i filer over tid.

Det gjør at man kan:

- se hva som er endret
- se hvem som gjorde endringen
- gå tilbake til tidligere versjoner
- samarbeide uten å overskrive hverandres arbeid



## Tidslinje

```text
Versjon 1        Versjon 2        Versjon 3        Versjon 4
Opprettet bok ─▶ Kapittel 1 ─▶ Rettet figur 2 ─▶ Lagt til slides
                                                        │
                                                        │ git push
                                                        ▼
                                                     GitHub
```


## Hvorfor trenger vi versjonskontroll?

> Versjonskontroll motvirker unødvendig redundans

```text
rapport.docx
rapport_ny.docx
rapport_ny_ny.docx
rapport_final.docx
rapport_final2.docx
rapport_final_final.docx
```

Etter en stund vet ingen hvilken fil som egentlig er den riktige.

Versjonskontroll løser dette ved å ha:

- én versjon av filene
- en historikk over alle endringer

# Gå tilbake til en tidligere versjon

En av de viktigste fordelene med versjonskontroll er at man kan hente frem tidligere versjoner av et prosjekt.

## Se historikken

For å se tidligere versjoner bruker man:

```bash
git log
```

Eksempel:

```text
a1b2c3d  Lagt til notat i rapport
e4f5g6h  Rettet figur 2
i7j8k9l  Opprettet kapittel 1
```

Hver commit representerer et øyeblikksbilde (*snapshot*) av prosjektet.

## Se en tidligere versjon

Hvis man ønsker å undersøke hvordan prosjektet så ut på et tidligere tidspunkt:

```bash
git checkout e4f5g6h
```

Da åpnes prosjektet slik det så ut da denne versjonen ble lagret.

For å gå tilbake til den nyeste versjonen:

```bash
git checkout main
```

## Gjenopprett en fil

Hvis en enkelt fil er blitt ødelagt eller slettet, kan den hentes fra en tidligere versjon:

```bash
git restore --source=e4f5g6h rapport.md
```

Dette gjenoppretter filen `rapport.md` fra den valgte commiten.




## Angre siste commit

Hvis den siste commiten ikke skal beholdes:

```bash
git reset --hard HEAD~1
```

Før:

```text
Versjon 1
Versjon 2
Versjon 3  ← HEAD
```

Etter:

```text
Versjon 1
Versjon 2  ← HEAD
```
> Kommandoen sletter lokale endringer.


## Kommandoer å kjenne til

For de fleste er det nok å kjenne til:

```bash
git log
git restore
git revert
```

Disse kommandoene gjør det mulig å finne, gjenopprette og angre tidligere endringer uten risiko for å miste hele prosjektet.

### Gå tilbake til tidligere versjoner
```text
rapport.docx
      │
      ▼

Versjonskontroll
      |
      ▼

v1 ── v2 ── v3 ── v4 ── v5
 │               │
 └───────────────┘
   kan gå tilbake
```


## Git-analogien

Git tar bilder (*snapshots*) av prosjektet underveis.

For eksempel:

```text
Versjon 1: Opprettet boka
Versjon 2: Lagt til kapittel 1
Versjon 3: Rettet figur 2
Versjon 4: Lagt til forelesningsslides
```

Man kan når som helst gå tilbake til en tidligere versjon.



## Eksempel

Legger vi til noe i Git:

```bash
git add .
git commit -m "Added something to repo"
git push
```

Dette betyr:

- `git add` velger hvilke endringer som skal lagres
- `git commit` oppretter en ny versjon
- `git push` sender versjonen til GitHub

Dermed blir alle endringer dokumentert og sporbare

## Fordeler

Versjonskontroll gjør det mulig å:

- unngå at arbeid går tapt
- eksperimentere uten å være redd for å ødelegge noe
- dokumentere utviklingen av et prosjekt
- samarbeide effektivt med andre
- publisere innhold automatisk gjennom verktøy som GitHub Actions



## Definisjon

> Versjonskontroll er systematisk lagring og sporing av endringer i filer, slik at man kan dokumentere, gjenfinne og samarbeide om utviklingen av et prosjekt.

## Grunnleggende Git-arbeidsflyt

For nybegynnere er det ofte nok å lære:

```text
clone
pull
add
commit
push
```

### Hva gjør `git pull`?

Kommandoen

```bash
git pull
```

henter de nyeste endringene fra GitHub og oppdaterer den lokale kopien av repoet.

Denne kommandoen brukes når andre har lagt til eller oppdatert filer i repoet, for eksempel nye datasett eller øvingsfiler.

For dette emnet er det i hovedsak disse to kommandoene som brukes:

```bash
git clone
git pull
```


## Opprette et Git-repo i Git Bash

### 1. Gå til ønsket mappe
Ved hjelp av *change directory* `cd` bytter man katalog:

```bash
cd /c/repos
```

### 2. Opprett en ny mappe
Lager ny katalog ved *make directory* `mkdir`:

```bash
mkdir dynamiske-kraftsystemer
```


### 3. Gå inn i mappen

```bash
cd dynamiske-kraftsystemer
```

### 4. Opprett Git-repo

```bash
git init
```

Du får en melding som ligner:

```text
Initialized empty Git repository
```




### 5. Opprett en README-fil

`echo` er en kommando som skriver tekst til terminalen:

Denne kommansoen skriver tekst til terminalen og deignerer til fil
```bash
echo "# Dynamiske kraftsystemer" > README.md
```

`cat`-kommando brukes til å sjekke innholdet
```bash
cat README.md
```
Output:
```text
# Dynamiske kraftsystemer
```
### 6. Legg filen til Git

```bash
git add README.md
```

### 7. Lag første commit

```bash
git commit -m "Initial commit"
```




### 8. Koble til GitHub (valgfritt)

Opprett først et tomt repository på GitHub, og koble deretter til:

```bash
git remote add origin https://github.com/brukernavn/dynamiske-kraftsystemer.git
```

### 9. Last opp til GitHub

Bytter navn på den nåværende branchen til main.

branch = jobb med brancher.
-M = tving omdøping (selv om main allerede finnes; master omdæpes til main).
main = nytt navn på branchen.

```bash
git branch -M main
```

Dette aster opp branchen main til GitHub.

push = send commits til GitHub.
origin = navnet på GitHub-repoet.
main = branchen som skal sendes.
-u = husk koblingen mellom lokal main og GitHubs main.

```bash
git push -u origin main
```

-u er en forkortelse for --set-upstream, og gjør at git husker koblingen mellom lokalt repo og repoet på GitHub.

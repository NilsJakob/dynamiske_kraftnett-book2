CLI er en måte å bruke et program på ved å skrive kommandoer i et tekstvindu i stedet for å klikke på knapper og menyer. I Fig. {numref}`fig-gw1` er et eksempel på kommandoer brukt i Git Bash. Målet var å lage en gratis og fritt tilgjengelig versjon av den opprinnelige Bourne Shell, den klassiske Unix-skallen (sh) utviklet av Stephen Bourne på 1970-tallet.

```{figure} ../images/gw1.png
:width: 80%
:align: center
:name: fig-gw1

Eksempel på kommandolinjegrensesnitt
```

##### Kuriositet: GNU og Linux

Målet var å bygge et komplett operativsystem, men GNU manglet én sentral komponent: kjernen (kernel).

I 1991 utviklet Linus Torvalds Linux-kjernen. Da kunne GNU-verktøyene kombineres med Linux-kjernen, og resultatet ble det vi i dag vanligvis kaller Linux.

Linux-kjernen ble utviklet av Linus Torvalds i 1991 og har pingvinen Tux (Torvalds UniX) som maskot, i Fig. {numref}`fig-tux`. Sammen med verktøy fra GNU-prosjektet danner Linux-kjernen grunnlaget for de fleste moderne Linux-distribusjoner.

```{figure} ../images/Linux_mascot_tux.png
:width: 250px
:align: center
:name: fig-tux

Tux, Linux-maskoten. Illustrasjon basert på originaldesignet til Larry Ewing (1996), tilgjengelig via Wikimedia Commons.
```

### 3. Start Git Bash

Etter installasjonen kan du:

- åpne **Git Bash** fra Start-menyen, eller
- høyreklikke i en mappe og velge **Open Git Bash Here**.



### 4. Test installasjonen

Skriv:

```bash
git --version
```

Eksempel på utskrift:

```text
git version 2.51.0.windows.1
```




### 5. Konfigurer navn og e-post

Git lagrer informasjon om hvem som oppretter commits. Derfor må navn og e-post registreres én gang:

```bash
git config --global user.name "Ditt Navn"
git config --global user.email "din.epost@example.com"
```

Disse opplysningene lagres sammen med alle commits og gjør det mulig å se hvem som har gjort ulike endringer i et prosjekt.



### 6. Klon et repo

```bash
git clone https://github.com/bruker/repo.git
```



### 7. Oppdater et repo
### Hva gjør `git pull`?

Kommandoen

```bash
git pull
```

henter de nyeste endringene fra GitHub og oppdaterer den lokale kopien av repoet.

Denne kommandoen brukes når andre har lagt til eller oppdatert filer i repoet, for eksempel nye datasett eller øvingsfiler.

For dette emnet er det i hovedsak disse to kommandoene som brukes:

```bash
git clone
git pull
```


## Opprette et Git-repo i Git Bash

### 1. Gå til ønsket mappe

```bash
cd /c/repos
```

### 2. Opprett en ny mappe

```bash
mkdir power-system-data
```


### 3. Gå inn i mappen

```bash
cd power-system-data
```

### 4. Opprett Git-repo

```bash
git init
```

Du får en melding som ligner:

```text
Initialized empty Git repository
```




### 5. Opprett en README-fil

```bash
echo "# Power System Data" > README.md
```

### 6. Legg filen til Git

```bash
git add README.md
```

### 7. Lag første commit

```bash
git commit -m "Initial commit"
```




### 8. Koble til GitHub (valgfritt)

Opprett først et tomt repository på GitHub, og koble deretter til:

```bash
git remote add origin https://github.com/brukernavn/power-system-data.git
```

### 9. Last opp til GitHub

```bash
git branch -M main
git push -u origin main
```



### Nyttige kommandoer

Vis status:

```bash
git status
```

Legg til alle endringer:

```bash
git add .
```

Lag commit:

```bash
git commit -m "Beskrivelse av endringen"
```

Last opp til GitHub:

```bash
git push
```

Hent siste endringer:

```bash
git pull
```

For å bruke funkjonen importerer vi python-biblioteket pandas som pd

```
import pandas as pd
```

Vi leser data fra csv filer med funksjoen:

```
pd.read_csv()
```

### Hvorfor bruker vi CSV-filer?

CSV står for comma seperated values. Hver verdi i en rad er separert med en delimiter. I dette tilfelle komma (vi kan også ha andre delimiter, men komma er mest vanlig)

I dette emnet brukes CSV-filer som standardformat for datasett.

En viktig grunn er at CSV-filer fungerer godt sammen med versjonskontrollsystemer som Git. En CSV-fil er en vanlig tekstfil, og Git kan derfor vise hvilke rader som er endret mellom ulike versjoner av datasettet.

For eksempel kan Git vise at en ny observasjon er lagt til:

```diff
2026-08-20 22:00,14064.84
2026-08-20 23:00,13594.91
+2026-08-21 00:00,13247.53
```



For å studere tallgrunnlaget kan vi lese rapporter fra pålitelige kilder som Norges Vassdrag og Energidirektorat (NVE), som rapporten: [Status og utvikling i kraftsystemet 2025-2030](https://publikasjoner.nve.no/rapport/2026/rapport2026_17.pdf), vist {numref}`fig-nve-rapport-2026-17`.

## Nedlastingsportaler for kraftsystemdata

For å gjøre det enkelt å komme i gang med dataanalyse bruker vi ferdig innsamlede datasett i CSV-format. Dataene er hentet fra åpne og anerkjente kilder i kraftsektoren.

### ENTSO-E Transparency Platform

ENTSO-E (European Network of Transmission System Operators for Electricity) publiserer data om blant annet:

- elektrisk last (forbruk),
- kraftproduksjon fordelt på energikilder,
- strømpriser,
- kraftutveksling mellom land,
- kapasitet og flaskehalser i nettet.

Disse dataene samles inn fra europeiske systemoperatører og er en viktig kilde for analyser av kraftsystemet. Data kan lastes ned som CSV-filer eller hentes direkte gjennom et API.

**Nettside:** https://www.entsoe.eu/data/

### Statnett

Statnett publiserer løpende driftsdata fra det norske kraftsystemet, blant annet:

- produksjon,
- forbruk,
- import og eksport,
- frekvens,
- kraftflyt mellom prisområder.

Dataene oppdateres fortløpende og gir innsikt i hvordan det norske kraftsystemet opererer i sanntid.

**Nettside:** https://www.statnett.no/

### Elhub

Elhub er den nasjonale datahuben for måledata fra norske strømkunder. Elhub samler inn og administrerer måleverdier fra smarte strømmålere (AMS) og brukes av nettselskaper, kraftleverandører og andre aktører i kraftmarkedet.

Elhub tilbyr blant annet:

- anonymiserte forbruksdata,
- statistikk om strømforbruk,
- data om målepunkter,
- innsikt i forbruksmønstre.

Elhub er en særlig viktig kilde for analyser av sluttbrukernes strømforbruk og elektrifisering.

**Nettside:** https://elhub.no/



### Statistisk sentralbyrå (SSB)

Statistisk sentralbyrå publiserer omfattende energi- og kraftstatistikk for Norge.

SSB er en viktig kilde for:

- energibalanser,
- elektrisitetsforbruk,
- kraftproduksjon,
- næringsstatistikk,
- befolknings- og samfunnsdata som kan brukes til å forklare energibruk.

Dataseriene er ofte tilgjengelige gjennom SSBs statistikkbank og kan lastes ned i ulike formater.

**Nettside:** https://www.ssb.no/

## Hente og analysere lastdata fra ENTSO-E

Dette eksemplet viser hvordan vi kan hente elektrisk last fra ENTSO-E Transparency Platform og visualisere forbruket for det siste komplette døgnet.

### Import av biblioteker

```python
import pandas as pd
from entsoe import EntsoePandasClient
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime
import time
```

Bibliotekene brukes til å:
- lese fra fil
- kommunisere med ENTSO-E,
- behandle tidsserier,
- lage figurer,
- registrere tidspunkt og responstid.

### Lese API-nøkkel

```python
load_dotenv("../.env")
```

API-nøkkelen lagres i en separat `.env`-fil slik at sensitive opplysninger ikke plasseres direkte i programkoden.

```python
client = EntsoePandasClient(
    api_key=os.getenv("ENTSOE_API_KEY")
)
```

Her opprettes en klient som brukes til å kommunisere med ENTSO-E.

### Velge tidsintervall

```python
today = pd.Timestamp.now(tz="Europe/Oslo").normalize()
start = today - pd.Timedelta(days=1)
end = today
```

Vi velger det siste komplette døgnet:

- `today` er dagens dato ved midnatt.
- `start` er midnatt dagen før.
- `end` er dagens midnatt.

### Hente lastdata

```python
load = client.query_load(
    country_code="NO",
    start=start,
    end=end
)
```

Her hentes faktisk elektrisk belastning for Norge.

Resultatet er en tidsserie der:

- indeksen inneholder tidspunkter,
- verdiene representerer last i MW.

### Konvertere til timesverdier

```python
load_hourly = load.resample("h").mean()
```

ENTSO-E leverer ofte data med høyere tidsoppløsning enn én time.

Ved hjelp av `resample("h")` beregnes gjennomsnittlig last for hver time.

### Visualisering

```python
plt.figure(figsize=(10,4))

plt.plot(
    load_hourly.index.hour,
    load_hourly.values,
    marker="o"
)

plt.title("Faktisk last siste komplette døgn")
plt.xlabel("Time")
plt.ylabel("Last (MW)")
plt.xticks(range(24))
plt.grid(True)

plt.show()
```

Figuren viser hvordan belastningen varierer gjennom døgnet.

Typiske trekk er:

- lav belastning om natten,
- økende belastning om morgenen,
- høy belastning på dagtid,
- reduksjon utover kvelden.

### Registrere tidspunkt og responstid

```python
request_time = datetime.now()

tic = time.perf_counter()

toc = time.perf_counter()
```

Disse variablene brukes for å dokumentere når analysen ble utført og hvor lang tid forespørselen tok.

### Skrive ut informasjon om datasettet

```python
print(f"API forespørsel sendt: {request_time:%Y-%m-%d %H:%M:%S}")
print(f"Forespurt intervall: {start} -> {end}")
print(f"Første målepunkt: {load.index.min()}")
print(f"Siste målepunkt: {load.index.max()}")
print(f"Analyserer siste komplette døgn: {start.strftime('%Y-%m-%d')}")
print(f"Antall målepunkter: {len(load)}")
print(f"Hentet {len(load)} målepunkter")
print(f"Responstid: {toc - tic:.2f} sekunder")
```

Denne informasjonen gir:

- tidspunkt for API-kallet,
- analysert tidsperiode,
- første og siste observasjon,
- antall målepunkter,
- responstid.

### Oppsummering

Programmet utfører følgende steg:

1. Leser inn API-nøkkel.
2. Oppretter forbindelse til ENTSO-E.
3. Henter lastdata for Norge.
4. Konverterer dataene til timesverdier.
5. Plotter døgnprofilen.
6. Skriver ut informasjon om datasettet.

Dette er et godt eksempel på en komplett dataanalyseprosess der vi:

```text
ENTSO-E
    ↓
Rådata
    ↓
Pandas
    ↓
Tidsserieanalyse
    ↓
Visualisering
```

Senere kan de samme dataene brukes til statistiske analyser, regresjon, prognoser og modellering av elektrisk belastning.

````python
# TEST CELL
from dotenv import load_dotenv
import os
import pandas as pd
from entsoe import EntsoePandasClient
import numpy as np
import matplotlib.pyplot as plt
from datetime import datetime
import time

load_dotenv("../.env")

client = EntsoePandasClient(
    api_key=os.getenv("ENTSOE_API_KEY")
)

today = pd.Timestamp.now(tz="Europe/Oslo").normalize()
start = today - pd.Timedelta(days=1)
end = today

load = client.query_load(
    country_code="NO",
    start=start,
    end=end
)

load_hourly = load.resample("h").mean()

plt.figure(figsize=(10,4))
plt.plot(load_hourly.index.hour,
         load_hourly.values,
         marker="o")

plt.title("Faktisk last siste komplette døgn")
plt.xlabel("Time")
plt.ylabel("Last (MW)")
plt.xticks(range(24))
plt.grid(True)
plt.show()

request_time = datetime.now()
tic = time.perf_counter()

toc = time.perf_counter()

print(f"API forespørsel sendt: {request_time:%Y-%m-%d %H:%M:%S}")
print(f"Forespurt intervall: {start} -> {end}")
print(f"Første målepunkt: {load.index.min()}")
print(f"Siste målepunkt: {load.index.max()}")
print(f"Analyserer siste komplette døgn: "f"{start.strftime('%Y-%m-%d')}")
print(f"Antall målepunkter: {len(load)}")
print(f"Hentet {len(load)} målepunkter")
print(f"Responstid: {toc - tic:.2f} sekunder")
#print(load_hourly)
````

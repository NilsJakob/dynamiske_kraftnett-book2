````python
# TEST CELL
import pandapower as pp

net = pp.create_empty_network(sn_mva=100)

b1 = pp.create_bus(net, vn_kv=20)
b2 = pp.create_bus(net, vn_kv=20)

pp.create_ext_grid(net, b1, vm_pu=1.0)

pp.create_impedance(
    net,
    from_bus=b1,
    to_bus=b2,
    rft_pu=0.1,
    xft_pu=0.5,
    sn_mva=100
)

pp.create_load(
    net,
    bus=b2,
    p_mw=30,
    q_mvar=20
)

pp.runpp(
net,
algorithm="gs",
max_iteration=30,
init="flat",
tolerance_mva=1e-6
) 

print(net.res_bus)
print(net._ppc["internal"].keys())
pp.runpp(
net,
algorithm="nr",
max_iteration=9,
init="flat",
tolerance_mva=1e-6,
v_debug=True
)

print(net.res_bus)
print(net._ppc["internal"].keys())
````

````python
pp.runpp(net, algorithm="nr", v_debug=True)

print(net._ppc["internal"]["Vm_it"])
print(net._ppc["internal"]["Va_it"])
````

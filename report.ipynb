ELK330 – Retningslinjer for rapportomfang

OMFANG

Rapporten skal dokumentere et selvstendig prosjektarbeid innen elektriske anlegg og installasjoner.

Arbeidet forventes å tilsvare ca. 40–60 arbeidstimer per student, inkludert:

- planlegging av prosjektet
- modellering av kraftsystemet
- simuleringer og beregninger
- analyse og vurdering av resultater
- rapportskriving og dokumentasjon

For gruppeprosjekter forventes arbeidsinnsatsen å være tilsvarende summen av antall gruppemedlemmer.

Rapporten vil normalt være på 15–25 sider, eksklusive:

- forside
- innholdsfortegnelse
- referanseliste
- vedlegg

KRAV TIL INNHOLD

Rapporten skal som minimum inneholde:

1. Problemformulering og målsetting
2. Beskrivelse av nettmodell og forutsetninger
3. Relevante beregninger og simuleringer
4. Presentasjon av resultater i tabeller og figurer
5. Analyse og diskusjon av resultatene
6. Kritisk vurdering av modellforutsetninger og begrensninger
7. Konklusjon
8. Referanser

LEVERANSER

Følgende skal leveres i Canvas:

- Rapport i PDF-format
- Simuleringsmodell (PowerFactory)
- Eventuelle beregningsfiler, skript eller støttedokumenter
- Vedlegg som dokumenterer sentrale modelleringsvalg og resultater

VURDERINGSKRITERIER

Det legges hovedvekt på:

- faglig korrekthet
- kvalitet på modellering og simulering
- analyse og tolkning av resultater
- teknisk refleksjon og kritisk vurdering
- struktur og kvalitet på rapporten

Det er kvaliteten på det faglige arbeidet som vurderes. Sideantall alene vil ikke være avgjørende for karakteren.

# ELK330 – Krav til prosjektrapport

## Generelt

Prosjektrapporten utgjør **2/5 av samlet karaktergrunnlag** i emnet ELK330.

Rapporten skal dokumentere prosjektarbeidet og studentenes evne til å modellere, simulere, analysere og reflektere over elektriske anlegg og installasjoner.

Det spesifiseres ikke et bestemt antall sider. Rapporten skal ha et omfang som er nødvendig for å dokumentere arbeidet på en faglig forsvarlig måte.

---

# Formelle krav

Studentene skal demonstrere:

- Beherskelse av Markdown-formattering (`.md`).
- Strukturert og lesbar dokumentasjon ved hjelp av overskrifter, figurer, tabeller og lignende.
- Korrekt bruk av hyperlenker og interne referanser.
- Korrekt kildebruk og akademisk referering.
- Bruk av BibTeX som grunnlag for referanseliste.
- Referanseliste formatert etter APA-standarden.

---

# Kildearbeid

Rapporten skal dokumentere gjennomført litteratursøk.

Dette innebærer:

- Identifisering av relevante faglige kilder.
- Beskrivelse av litteratursøket.
- Kritisk vurdering av kildenes relevans.
- Bruk av fagfellevurderte artikler, standarder, lærebøker og teknisk dokumentasjon der dette er relevant.

---

# Versjonskontroll og reproduksjon

Det skal leveres:

- Lenke til Git-repositorium.
- Git-logg som dokumenterer utviklingen av prosjektet gjennom semesteret.

Versjonshistorikken skal vise aktiv bruk av Git som verktøy for dokumentasjon og utvikling.

---

# Modellering

Studentene skal gjøre rede for:

- Oppbygningen av simuleringsmodellen.
- Valgte forutsetninger.
- Forenklinger og avgrensninger.
- Parametervalg.

Rapporten skal diskutere modellens:

- Underrepresentasjon av virkeligheten.
- Overrepresentasjon av virkeligheten.

Diskusjonen skal knyttes til emnets teori om modeller og modellering.

---

# Simulering

Følgende skal leveres:

- PowerFactory-modell (`.pfd`).
- Eventuelle tilhørende datafiler.

Rapporten skal beskrive:

- Gjennomførte simuleringer.
- Valgte scenarioer.
- Resultater og observasjoner.

---

# Analyse og refleksjon

Det forventes en selvstendig faglig diskusjon av resultatene.

Studentene skal:

- Tolke simuleringsresultatene.
- Vurdere resultatets usikkerhet og begrensninger.
- Diskutere hvilke forhold modellen fanger opp og hvilke forhold modellen ikke fanger opp.
- Knytte analysen til emnets teori om digitale modeller.

Studentene skal videre identifisere modellens plassering med hensyn til:

- Integrasjon.
- Kompleksitet.

Diskusjonen skal relateres til begrepene introdusert i emnets kapittel om digitale modeller.

---

# Vurderingskriterier

Rapporten vurderes ut fra:

1. Faglig kvalitet i modellering og simulering.
2. Evne til analyse, refleksjon og kritisk vurdering.
3. Kobling mellom teori, modell og simuleringsresultater.
4. Kvalitet på dokumentasjon og rapportstruktur.
5. Kvalitet på litteratursøk og kildebruk.
6. Bruk av Git og dokumentasjon av arbeidsprosessen.
7. Reproduserbarhet gjennom vedlagte modeller og prosjektfiler.

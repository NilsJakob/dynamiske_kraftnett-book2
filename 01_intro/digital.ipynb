# Digitalisering
## Kraftsystemdata
Utviklingen viser at storskala kraftoverføring i høyspentnettet, energilagring og den digitale revolusjonen er sentrale drivkrefter for en omfattende avkarbonisering av kraftsektoren, fremveksten av nye markeds- og forretningsmodeller på distribusjonsnivå samt utviklingen av nye arkitekturer for kraftdistribusjon.

Eksperter innen kraftsystemanalyse peker gjerne på de såkalte [3D'ene](https://www.sciencedirect.com/science/article/pii/S1364032118304283) (dekarbonisering, desentralisering og digitalisering) som sentrale drivkrefter bak endringene i kraftsystemet. Disse og andre D-rammeverk ([D'frames](https://www.sciencedirect.com/science/article/pii/S2214629621003236#b0140)) brukes som analytiske verktøy for å forstå trendene som former framtidens kraftnett.

Kraftsystemdata beskriver og dokumenterer de underliggende prosessene som driver denne utviklingen. Data fra målesystemer, sensorer, markedsplattformer og nettstyringssystemer gir innsikt i hvordan produksjon, forbruk og effektflyt endrer seg over tid.

Et sentralt dilemma er imidlertid at avanserte [ki-algoritmer som skal bidra til å opitmalisere strømbruken](https://www.europower.no/digitalisering/ingen-kunstig-intelligens-uten-barekraft-slik-kan-vi-toyle-stromforbruket/2-1-1665594), samtidig kan ha et betydelig energiforbruk. 

Utviklingen av framtidens digitale kraftsystem krever derfor en balansegang mellom økt beregningskraft, energieffektivitet og bærekraft. Digitaliseringen representerer dermed både en mulig løsning på og en ny utfordring for energitrilemmaet. [Energitrilemmaet](https://brandcentral.ramboll.com/share/xRsGFgZS52Xq4iMi3RcJ/assets/69338) beskriver utfordringen med samtidig å sikre høy forsyningssikkerhet, lavest mulig miljø- og klimabelastning samt økonomisk tilgjengelig energi. Dette samsvarer i stor grad med målsettingene i FNs bærekraftsmål 7 (Sustainable Development Goal - [SDG](https://sdgs.un.org/goals)7), som vektlegger tilgang til pålitelig, bærekraftig og moderne energi til en overkommelig pris. Avanserte analyseverktøy og kunstig intelligens kan bidra til økt fleksibilitet og bedre ressursutnyttelse i kraftsystemet, men samtidig øker energibehovet til den digitale infrastrukturen som skal understøtte denne utviklingen.

I norsk energidebatt har NVE-direktør Ketil Lund beskrevet den såkalte [strømskvisen](https://www.vg.no/nyheter/i/pQ6y6G/den-store-stroemskvisen) som et trilemma mellom grønn omstilling, bevaring av natur og rimelig energi. Trilemmaet illustrerer at tiltak som fremmer ett av målene, ofte kan komme i konflikt med de to øvrige. Digitaliseringen av kraftsystemet må derfor ikke bare vurderes ut fra teknologiske muligheter, men også i lys av hvordan den påvirker klima, natur og energikostnader.

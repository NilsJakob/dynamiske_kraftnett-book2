 # Jupyter Bok

[Jupyter Book](https://jupyterbook.org/) er et verktøy med åpen kildekode som brukes til å bygge bøker, nettsider og dokumenter fra en samling interaktive notatbøker (jupyter notebook .ipynb) og Markdown-filer (.md). Prosjektet blir hostet, utviklet og vedlikeholdt som et åpen kildekode-prosjekt på GitHub. En slik notatbok er Jupyter Notebook. Jupyter Notebook er et gratis, nettbasert verktøy som støtter Python og lar deg skrive og kjøre kode i interaktive celler, legge til forklarende tekst og inkludere hyperlenker i Markdown-celler, slik at du kan arbeide med data, kode og dokumentasjon på én og samme side.

[GitHub](https://github.com/) er en skybasert plattform og nettside som brukes til å administrere kode. Plattformen bygger på Git, som er verktøyet som kjøres lokalt på datamaskinen og registrerer historikken til filer og prosjekter. GitHub gjør det mulig for utviklere å samarbeide om prosjekter, spore endringer og bevare historikken til alle versjoner av koden. Plattformen brukes i stor utstrekning av enkeltutviklere, team og [organisasjoner](https://github.com/) til utvikling av programvare.

### Hvordan opprette en GitHub-konto

1. Gå til https://github.com
2. Klikk på **"Sign up"**.
3. Oppgi:
   - E-postadresse
   - Passord
   - Brukernavn (må være unikt)
4. Bekreft e-postadressen når GitHub sender en bekreftelsesmelding.
5. Fullfør eventuelle verifiseringstrinn.
6. Velg **Free**-abonnementet.
7. Fullfør oppsettveiviseren.

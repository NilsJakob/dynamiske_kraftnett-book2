## Steg 3: Bestem ligningene for aktiv og reaktiv effekt

Den generelle effektligningen er

$$
P_i
=
\sum_{j=1}^{n}
|V_i||V_j||Y_{ij}|
\cos(\theta_{ij}+\delta_j-\delta_i)
$$

$$
Q_i
=
-
\sum_{j=1}^{n}
|V_i||V_j||Y_{ij}|
\sin(\theta_{ij}+\delta_j-\delta_i)
$$

For buss 2:

$$
P_2
=
|V_2||V_1||Y_{21}|
\cos(\theta_{21}+\delta_1-\delta_2)
+
|V_2|^2|Y_{22}|
\cos(\theta_{22})
$$

$$
Q_2
=
-
|V_2||V_1||Y_{21}|
\sin(\theta_{21}+\delta_1-\delta_2)
-
|V_2|^2|Y_{22}|
\sin(\theta_{22})
$$

Setter inn verdiene:

$$
P_2
=
|V_2|
\cdot 1.0
\cdot 9.7529
\cos(105.46^\circ-\delta_2)
+
|V_2|^2
\cdot 9.7529
\cos(-74.53^\circ)
$$

$$
Q_2
=
-
|V_2|
\cdot 1.0
\cdot 9.7529
\sin(105.46^\circ-\delta_2)
-
|V_2|^2
\cdot 9.7529
\sin(-74.53^\circ)
$$

## Steg 9: Jacobimatrisen og korreksjonsvektoren

Jacobimatrisen er

$$
J=
\begin{bmatrix}
\frac{\partial P_2}{\partial \delta_2}
&
\frac{\partial P_2}{\partial |V_2|}
\\[8pt]
\frac{\partial Q_2}{\partial \delta_2}
&
\frac{\partial Q_2}{\partial |V_2|}
\end{bmatrix}
$$

Ved flat start:

$$
J=
\begin{bmatrix}
9.4000 & 2.6000\\
-2.6000 & 9.4000
\end{bmatrix}
$$

Mismatch-vektoren fra Steg 6:

$$
\Delta S=
\begin{bmatrix}
\Delta P_2^{(1)}\\
\Delta Q_2^{(1)}
\end{bmatrix}
=
\begin{bmatrix}
-1.5000\\
-0.5000
\end{bmatrix}
$$

Newton-Raphson-ligningen:

$$
\begin{bmatrix}
\Delta\delta_2^{(0)}\\
\Delta |V_2|^{(0)}
\end{bmatrix}
=
J^{-1}
\begin{bmatrix}
\Delta P_2^{(1)}\\
\Delta Q_2^{(1)}
\end{bmatrix}
$$

Den inverse Jacobimatrisen er

$$
J^{-1}
=
\begin{bmatrix}
0.0988 & -0.0273\\
0.0273 & 0.0988
\end{bmatrix}

$$
\begin{bmatrix}
\Delta\delta_2^{(0)}\\

-0.0904
\end{bmatrix}

eller

=

$$

\ \text{pu}

\angle
(-7.7101^\circ)

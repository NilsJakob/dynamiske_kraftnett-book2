# Oppgave: Økt belastning og spenningsstabilitet

## Bakgrunn

I denne oppgaven skal du undersøke hvordan et kraftsystem oppfører seg når lasten gradvis økes. Målet er å studere sammenhengen mellom belastning, spenning, strøm, effekttap og spenningsstabilitet.

Når lasten øker, må mer effekt transporteres gjennom linjen. Dette fører til økt strøm, større spenningsfall og økte tap. Dersom belastningen blir for høy, kan systemet nærme seg en tilstand der det ikke lenger klarer å opprettholde tilfredsstillende spenning. Dette fenomenet kalles **spenningsstabilitetsproblemer** eller **spenningskollaps**.

---

## Systembeskrivelse

Bruk samme 2-buss-system som i Laboratorium 1.

```text
      Bus 1 (Slack)
      V₁ = 1∠0°
           |
           |
      Z = 0.1+j0.5 pu
           |
           |
      Bus 2 (PQ)
      P = Last
      Q = -0.2 pu
```

\[

## Oppgave 5: Spenningskollaps

Studer resultatene dine og vurder om systemet nærmer seg spenningskollaps.

### Teori

Når belastningen øker:

- Busspenningen blir stadig lavere.
- Linjestrømmen blir stadig høyere.
- Effekttapene øker kraftig.
- Lastflytberegningen kan få problemer med å konvergere.

Dette er typiske tegn på at systemet nærmer seg en spenningskritisk tilstand.

4. Hvorfor blir spenningsstabilitet viktigere i svake nett enn i sterke nett?

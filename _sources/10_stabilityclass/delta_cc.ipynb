# Kritisk utkoblingsvinkel $\delta_{cc}$ ved Equal Area Criterion

Lik areal-betingelsen gir

$$
A_1 = A_2
$$

der

- $A_1$ = akselerasjonsarealet under feilen
- $A_2$ = retardasjonsarealet etter at feilen er koblet ut

---

## Initial og post-fault kraftvinkler

Før feilen:

$$
P_{m1}\sin\delta_0 = P_s
$$

Etter feilen:

$$
P_{m3}\sin\delta_s = P_s
$$

---

## Areal $A_1$

Fra figuren:

$$
A_1
=
P_s(\delta_{cc}-\delta_0)
-
\int_{\delta_0}^{\delta_{cc}}
P_{m2}\sin\delta \, d\delta
$$

Integrasjon:

$$
\int \sin\delta\,d\delta
=
-\cos\delta
$$

gir

$$
A_1
=
P_s(\delta_{cc}-\delta_0)
+
P_{m2}\cos\delta_{cc}
-
P_{m2}\cos\delta_0
$$

eller

$$
A_1
=
P_s\delta_{cc}
-
P_s\delta_0
+
P_{m2}\cos\delta_{cc}
-
P_{m2}\cos\delta_0
$$

---

## Areal $A_2$

Fra figuren:

$$
A_2
=
\int_{\delta_{cc}}^{\delta_m}
P_{m3}\sin\delta\,d\delta
-
P_s(\delta_m-\delta_{cc})
$$

Integrasjon gir

$$
A_2
=
P_{m3}\cos\delta_{cc}
-
P_{m3}\cos\delta_m
-
P_s\delta_m
+
P_s\delta_{cc}
$$

---

## Equal Area Criterion

Ved grensestabilitet:

$$
A_1=A_2
$$

Dermed:

$$
P_s\delta_{cc}
-
P_s\delta_0
+
P_{m2}\cos\delta_{cc}
-
P_{m2}\cos\delta_0
=
P_{m3}\cos\delta_{cc}
-
P_{m3}\cos\delta_m
-
P_s\delta_m
+
P_s\delta_{cc}
$$

Kansellerer leddene $P_s\delta_{cc}$:

$$
-P_s\delta_0
+
P_{m2}\cos\delta_{cc}
-
P_{m2}\cos\delta_0
=
P_{m3}\cos\delta_{cc}
-
P_{m3}\cos\delta_m
-
P_s\delta_m
$$

Samler leddene med $\cos\delta_{cc}$:

$$
(P_{m2}-P_{m3})
\cos\delta_{cc}
=
P_{m2}\cos\delta_0
-
P_{m3}\cos\delta_m
+
P_s(\delta_0-\delta_m)
$$

---

## Uttrykk for kritisk utkoblingsvinkel

$$
\boxed{
\cos\delta_{cc}
=
\frac{
P_{m2}\cos\delta_0
-
P_{m3}\cos\delta_m
+
P_s(\delta_0-\delta_m)
}{
P_{m2}-P_{m3}
}
}
$$

og dermed

$$
\boxed{
\delta_{cc}
=
\cos^{-1}
\left(
\frac{
P_{m2}\cos\delta_0
-
P_{m3}\cos\delta_m
+
P_s(\delta_0-\delta_m)
}{
P_{m2}-P_{m3}
}
\right)
}
$$

---

## Maksimal rotorvinkel

Etter feilen gjelder

$$
P_e=P_{m3}\sin\delta
$$

Likevektspunktet etter feilutkobling finnes fra

$$
P_s=P_{m3}\sin\delta_s
$$

Siden

$$
\sin\delta_s=\sin(\pi-\delta_s)
$$

fås et nytt skjæringspunkt på den fallende delen av sinuskurven:

$$
\boxed{
\delta_m=\pi-\delta_s
}
$$

Dette er den maksimale rotorvinkelen ved grensestabilitet.

---

## Fremgangsmåte

1. Finn initial vinkel:

$$
\delta_0
=
\sin^{-1}\left(\frac{P_s}{P_{m1}}\right)
$$

2. Finn post-fault likevektsvinkel:

$$
\delta_s
=
\sin^{-1}\left(\frac{P_s}{P_{m3}}\right)
$$

3. Beregn maksimal vinkel:

$$
\delta_m=\pi-\delta_s
$$

4. Beregn kritisk utkoblingsvinkel:

$$
\delta_{cc}
=
\cos^{-1}
\left(
\frac{
P_{m2}\cos\delta_0
-
P_{m3}\cos\delta_m
+
P_s(\delta_0-\delta_m)
}{
P_{m2}-P_{m3}
}
\right)
$$

5. Bruk swing-ligningen for å finne tilhørende kritisk utkoblingstid $t_{cc}$ dersom ønskelig.

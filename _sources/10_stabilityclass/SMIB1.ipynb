# BASERT PÅ KOTHARI

# Generell metode for transientstabilitetsanalyse av SMIB-systemer

## Formål

Metoden bygger på én enkel hovedregel:

> Alle feil modelleres ved å innføre en feilkoblingsbuss i feilpunktet.

Dermed kan samme framgangsmåte benyttes for:

- håndberegninger,
- nodeeliminasjon,
- Y-buss-metoder,
- like-areal-kriteriet,
- PowerFactory-simuleringer.

---

# Trinn 1: Modellér prefault-nettverket

Bestem systemets ekvivalente reaktans mellom generatorens interne spenning og infinite bus:

$$
X_{\text{pre}}
=
X_d' + X_t + X_{nett}
$$

Maksimal overførbar effekt blir

$$
P_{\max,\text{pre}}
=
\frac{E'V}{X_{\text{pre}}}
$$

og effektvinkelkurven

$$
P_{e,\text{pre}}
=
P_{\max,\text{pre}}\sin\delta
$$

---

# Trinn 2: Innfør feilkoblingsbuss

Dersom feilen ligger på en linje, deles linjen i to deler.

Før:

```text
Bus 1 -------- Bus 2
         x
```

Etter:

```text
Bus 1 --- Bus F --- Bus 2
             |
           Fault
```

---

# Trinn 3: Modellér feiltilstanden

For en 3-fase kortslutning:

$$
Z_f \approx 0
$$

Opprett nytt reaktansdiagram.

Reduser deretter nettet mellom

$$
E'
\quad\text{og}\quad
V
$$

for å finne

$$
X_{\text{fault}}
$$

Maksimal overførbar effekt under feilen:

$$
P_{\max,\text{fault}}
=
\frac{E'V}{X_{\text{fault}}}
$$

---

# Trinn 4: Modellér feilklarering

Etter at feilen fjernes:

- enten gjenopprettes opprinnelig nett,
- eller den feilrammede linjen kobles ut.

Finn da

$$
X_{\text{post}}
$$

og

$$
P_{\max,\text{post}}
=
\frac{E'V}{X_{\text{post}}}
$$

slik at

$$
P_{e,\text{post}}
=
P_{\max,\text{post}}\sin\delta
$$

---

# Trinn 5: Analyser stabilitet

Startvinkelen finnes fra

$$
P_m
=
P_{\max,\text{pre}}
\sin\delta_0
$$

slik at

$$
\delta_0
=
\sin^{-1}
\left(
\frac{P_m}
     {P_{\max,\text{pre}}}
\right)
$$

Bruk deretter:

- like-areal-kriteriet, eller
- svingligningen

for å bestemme

$$
\delta_{cr}
$$

og eventuelt

$$
t_{cr}
$$

---

# Flytskjema

```text
                 START
                    |
                    v
       Opprett prefault-nettverk
                    |
                    v
              Finn Xpre
                    |
                    v
           Beregn Pmax,pre
                    |
                    v
             Oppstår feil?
                    |
                    v
        Opprett feilkoblingsbuss
                    |
                    v
        Legg inn kortslutning
                    |
                    v
          Reduser feilnettverk
                    |
                    v
             Finn Xfault
                    |
                    v
          Beregn Pmax,fault
                    |
                    v
           Klareres feilen?
                    |
                    v
                  Ja
                    |
                    v
      Oppdater postfault-nettverk
                    |
                    v
              Finn Xpost
                    |
                    v
           Beregn Pmax,post
                    |
                    v
        Like-areal-kriterium
          eller simulering
                    |
                    v
       Bestem stabilitet
                    |
                    v
                  SLUTT
```

---

# BASERT PÅ GLOVER AND OVERBYE EXAMPLE 11.3

# Generell metode for transientstabilitetsanalyse av SMIB-systemer

## Formål

Målet er å etablere én generell metode som kan brukes på alle SMIB-problemer:

- feil på en buss,
- feil i enden av en linje,
- feil midt på en linje,
- håndberegninger,
- Y-buss-metoder,
- nodeeliminasjon (Kron-reduksjon),
- PowerFactory-simuleringer.

Metoden bygger på følgende hovedregel:

> Alle feil modelleres ved å innføre en feilkoblingsbuss (Fault Bus) i feilpunktet.

Dermed brukes samme framgangsmåte uansett problemtype.

---

# Grunnmodell

Vi betrakter en synkrongenerator koblet til et uendelig nett gjennom en transformator og et transmisjonsnett.

```text
E' ── Xd' ── Xt ── Nettverk ── V∞
```

der

- \(E'\) = intern generatorspenning
- \(X_d'\) = transientreaktans
- \(X_t\) = transformatorreaktans
- \(V_\infty\) = infinite bus

---

# Trinn 1: Prefault-analyse

Reduser nettverket mellom generatorens interne spenning og infinite bus.

Bestem den ekvivalente reaktansen

$$
X_{\text{pre}}
$$

Maksimal overførbar effekt er

$$
P_{\max,\text{pre}}
=
\frac{E'V_\infty}{X_{\text{pre}}}
$$

og effektvinkelkurven blir

$$
P_{e,\text{pre}}
=
P_{\max,\text{pre}}\sin\delta
$$

---

# Trinn 2: Innfør feilkoblingsbuss

Hvis feilen ligger på en buss:

- bruk eksisterende buss.

Hvis feilen ligger på en linje:

- del linjen i to deler,
- opprett en ny buss i feilpunktet.

Eksempel:

Før:

```text
Bus 1 -------- Bus 2
         ×
```

Etter:

```text
Bus 1 --- Bus F --- Bus 2
             |
          Fault
```

Dermed behandles alle feil på identisk måte.

---

# Trinn 3: Modellér feiltilstanden

For en trefase kortslutning:

$$
Z_f \approx 0
$$

Feilkoblingsbussen kobles mot jord:

```text
Bus F
  |
  |
 Ground
```

Reduser deretter nettet på nytt.

Bestem

$$
X_{\text{fault}}
$$

og

$$
P_{\max,\text{fault}}
=
\frac{E'V_\infty}{X_{\text{fault}}}
$$

slik at

$$
P_{e,\text{fault}}
=
P_{\max,\text{fault}}\sin\delta
$$

---

# Trinn 4: Modellér feilklarering

Når feilen fjernes, finnes to vanlige alternativer.

## Alternativ A

Feilen fjernes og systemet går tilbake til opprinnelig nett.

```text
Fault removed
```

---

## Alternativ B

Feilen fjernes og den feilrammede linjen kobles ut.

```text
Fault removed
Faulted line disconnected
```

Dette er den klassiske situasjonen i mange transientstabilitetsoppgaver.

---

# Trinn 5: Postfault-analyse

Reduser postfault-nettverket.

Bestem

$$
X_{\text{post}}
$$

og

$$
P_{\max,\text{post}}
=
\frac{E'V_\infty}{X_{\text{post}}}
$$

slik at

$$
P_{e,\text{post}}
=
P_{\max,\text{post}}\sin\delta
$$

---

# Trinn 6: Stabilitetsanalyse

Bestem startvinkelen

$$
\delta_0
=
\sin^{-1}
\left(
\frac{P_m}
{P_{\max,\text{pre}}}
\right)
$$

Deretter benyttes

- like-areal-kriteriet,
- svingligningen,
- eller tidsdomeneanalyse

for å bestemme stabiliteten.

---

# Flytskjema

```text
                 START
                    |
                    v
      Opprett prefault-nettverk
                    |
                    v
              Finn Xpre
                    |
                    v
          Beregn Pmax,pre
                    |
                    v
            Oppstår feil?
                    |
                    v
      Opprett feilkoblingsbuss
                    |
                    v
      Legg inn kortslutning
                    |
                    v
        Reduser feilnettverk
                    |
                    v
             Finn Xfault
                    |
                    v
         Beregn Pmax,fault
                    |
                    v
          Klareres feilen?
                    |
                    v
                  Ja
                    |
                    v
     Oppdater postfault-nettverk
                    |
                    v
             Finn Xpost
                    |
                    v
         Beregn Pmax,post
                    |
                    v
       Like-areal-kriterium
        eller simulering
                    |
                    v
      Bestem stabilitet
                    |
                    v
                  SLUTT
```

---

# GLOVER AND OVERBYE EXAMPLE 11.7


# Euler-metoden som stabilitetsanalyse i SMIB-systemer

## Bakgrunn

Tradisjonelt introduseres transientstabilitet gjennom like-areal-kriteriet og begrepene:

- kritisk ryddevinkel
- kritisk ryddetid

For bachelorstudenter kan dette imidlertid gi en bratt læringskurve fordi man samtidig må håndtere:

- lastflyt
- nettverksreduksjon
- effektvinkelkurver
- svingligningen
- arealbetraktninger

Et alternativ er å la studentene avgjøre stabilitet direkte ved hjelp av numerisk simulering av svingligningen med Euler-metoden.

Dette gir en arbeidsflyt som ligger nærmere moderne analyseverktøy som PowerFactory.

---

# Foreslått faglig progresjon

```text
Lastflyt
   ↓
Intern generatorspenning E'
   ↓
Nettverksreduksjon
(Kron-reduksjon)
   ↓
Prefault / Fault / Postfault
   ↓
Effektvinkelkurver
   ↓
Svingligningen
   ↓
Euler-metoden
   ↓
Stabilitetsvurdering
```

Studentene trenger da ikke å lære like-areal-kriteriet for å kunne analysere transient stabilitet.

---

# Svingligningen

Vi benytter den klassiske maskinmodellen

$$
\frac{d\delta}{dt}
=
\omega
$$

$$
\frac{d\omega}{dt}
=
\frac{\omega_s}{2H}
\left(
P_m-P_e
\right)
$$

der

- $\delta$ = rotorvinkel
- $\omega$ = relativ rotorhastighet
- $P_m$ = mekanisk effekt
- $P_e$ = elektrisk effekt
- $H$ = treghetskonstant
- $\omega_s$ = synkron vinkelhastighet

---

# Elektrisk effekt

Etter nettverksreduksjon kan den elektriske effekten uttrykkes som

$$
P_e
=
P_{\max}
\sin\delta
$$

der

$$
P_{\max}
=
\frac{E'V}{X}
$$

Det finnes normalt tre ulike effektvinkelkurver:

Prefault:

$$
P_{e,\text{pre}}
=
P_{\max,\text{pre}}
\sin\delta
$$

Under feil:

$$
P_{e,\text{fault}}
=
P_{\max,\text{fault}}
\sin\delta
$$

Postfault:

$$
P_{e,\text{post}}
=
P_{\max,\text{post}}
\sin\delta
$$

---

# Euler-metoden

En enkel eksplisitt Euler-tilnærming gir

$$
\delta_{k+1}
=
\delta_k
+
\Delta t\,\omega_k
$$

og

$$
\omega_{k+1}
=
\omega_k
+
\Delta t
\frac{\omega_s}{2H}
\left(
P_m-P_e(\delta_k)
\right)
$$

for hvert tidssteg.

---

# Stabilitet uten like-areal-kriteriet

I stedet for å beregne kritisk ryddevinkel analyseres den numeriske løsningen direkte.

---

## Stabil respons

Hvis rotorhastigheten begynner å synke etter feilklarering

$$
\omega \rightarrow 0
$$

og rotorvinkelen nærmer seg en ny likevekt,

$$
\delta
\rightarrow
\delta_{eq}
$$

anses systemet som stabilt.

Illustrasjon:

```text
ω

|
|\
| \
|  \
|   \____
|
+---------------> t
```

Rotoren akselererer først og bremses deretter tilbake.

---

## Ustabil respons

Hvis

$$
\omega
$$

forblir positiv,

og

$$
\delta
$$

fortsetter å øke,

oppnås ingen ny likevekt.

Illustrasjon:

```text
δ

|
|       /
|      /
|     /
|    /
|   /
+---------------> t
```

Rotoren mister synkronismen med nettet.

---

# Alternativ 1: Sammenligne to ryddetider

En enkel eksamensoppgave kan være:

## Tilfelle A

Feilen ryddes ved

$$
t_c = 0.15\,s
$$

## Tilfelle B

Feilen ryddes ved

$$
t_c = 0.20\,s
$$

Studentene utfører Euler-integrasjon for begge tilfellene.

Deretter vurderes:

- nærmer rotorhastigheten seg null?
- får vi en ny likevekt?
- eller fortsetter rotoren å akselerere?

På denne måten kan stabilitet avgjøres uten kritisk ryddevinkel.

---

# Alternativ 2: Starte ved feilklarering

Dette reduserer regnearbeidet betydelig.

I stedet for å integrere hele feilforløpet oppgis tilstanden ved feilklarering:

$$
\delta(t_c)=110^\circ
$$

$$
\omega(t_c)=0.45\,rad/s
$$

Studentene skal deretter beregne for eksempel 10 Euler-steg.

Dette tester:

- svingligningen
- Euler-metoden
- forståelse av stabilitet

uten at flere sider med manuelle beregninger kreves.

---

# Alternativ 3: Bruke akselerasjonseffekten

Definer

$$
P_a
=
P_m-P_e
$$

---

## Under feil

Vanligvis er

$$
P_m > P_e
$$

slik at

$$
P_a > 0
$$

Generatoren akselererer.

---

## Etter feilklarering

Hvis

$$
P_e > P_m
$$

oppnår vi

$$
P_a < 0
$$

Generatoren bremser.

---

# Tolkning

Studentene kan følge fortegnet til

$$
P_a
$$

for å forstå den fysiske utviklingen.

```text
Feil:

Pm > Pe
⇒ Pa > 0
⇒ Rotor akselererer

Etter feilklarering:

Pe > Pm
⇒ Pa < 0
⇒ Rotor bremses
```

Dermed oppnås en intuitiv forståelse av transient stabilitet.

---

# Forslag til eksamensstruktur

## Del A: Lastflyt

Bestem

$$
E'
$$

og strømmer i systemet.

---

## Del B: Nettverksreduksjon

Bestem

$$
X_{\text{pre}}
$$

$$
X_{\text{fault}}
$$

$$
X_{\text{post}}
$$

ved hjelp av serie-/parallellreduksjon eller Kron-reduksjon.

---

## Del C: Effektvinkelkurver

Bestem

$$
P_{\max,\text{pre}}
$$

$$
P_{\max,\text{fault}}
$$

$$
P_{\max,\text{post}}
$$

og skriv opp tilsvarende effektvinkeluttrykk.

---

## Del D: Euler-metoden

Løs svingligningen numerisk for et gitt antall tidssteg.

Beregn:

$$
\delta_k
$$

$$
\omega_k
$$

for hvert steg.

---

## Del E: Stabilitetsvurdering

Basert på utviklingen av:

$$
\delta(t)
$$

og

$$
\omega(t)
$$

avgjør om systemet ser ut til å være:

- stabilt
- marginalt stabilt
- ustabilt

---

# Pedagogisk fordel

Denne framgangsmåten gir én sammenhengende arbeidsflyt:

```text
Lastflyt
   ↓
Intern generatorspenning
   ↓
Kron-reduksjon
   ↓
Prefault/Fault/Postfault
   ↓
Effektvinkelkurver
   ↓
Svingligning
   ↓
Euler-metoden
   ↓
Stabilitetsvurdering
```

Studentene får dermed trening i:

- kraftsystemanalyse
- nettverksreduksjon
- numeriske metoder
- transient stabilitet

uten at like-areal-kriteriet introduseres i første omgang.

# KOTHARI TO-PORT-MODELL

# Fra lastflyt til transientstabilitet ved bruk av Kotharis to-port-modell

## Motivasjon

I mange lærebøker behandles følgende temaer som separate emner:

- lastflyt
- nettverksreduksjon
- transientstabilitet
- like-areal-kriteriet
- numerisk integrasjon

En alternativ tilnærming er å bruke Kotharis to-port-modell som en felles matematisk plattform gjennom hele emnet.

Da oppnås følgende faglige progresjon:

```text
Lastflyt
   ↓
Intern generatorspenning E'
   ↓
Y-buss
   ↓
Kron-reduksjon
   ↓
To-port-modell
   ↓
Effektvinkelkurve
   ↓
Svingligning
   ↓
Euler-metoden
   ↓
Stabilitetsanalyse
```

Denne arbeidsflyten ligger også nær hvordan moderne verktøy som PowerFactory gjennomfører transientstabilitetsberegninger.

# Kotharis stabilitetsmodell

Ved stabilitetsstudier erstattes det opprinnelige nettet med en redusert modell.

Alle laster antas representert som ekvivalente admittanser, og hele nettet reduseres til en to-port mellom maskinen og resten av systemet.

```text
E1'  ────── Stabilitetsnettverk ────── E2'
```

For et SMIB-system representerer

$$
E_1' = E'
$$

generatorens interne spenning, mens

$$
E_2' = V_\infty
$$

representerer infinite bus.

---

# To-port-modellen

Etter Kron-reduksjon kan systemet beskrives med

$$
\mathbf{Y}_{BUS}
=
\begin{bmatrix}
Y_{11} & Y_{12}\\
Y_{21} & Y_{22}
\end{bmatrix}
$$

der

$$
Y_{12}=Y_{21}
$$

for et symmetrisk nett.

Kompleks effekt ved buss 1 skrives

$$
S_1
=
P_1+jQ_1
=
E_1'I_1^{*}
$$

---

# Effektuttrykket

Kothari viser at aktiv effekt ved buss 1 kan uttrykkes som

$$
P_1
=
|E_1'|^2 G_{11}
+
|E_1'||E_2'||Y_{12}|
\cos(\delta_1-\delta_2-\theta_{12})
$$

der

$$
Y_{12}
=
|Y_{12}|\angle\theta_{12}
$$

og

$$
G_{11}
=
\operatorname{Re}(Y_{11})
$$

---

# Generell effektvinkelkurve

Definer

$$
P_c
=
|E_1'|^2G_{11}
$$

og

$$
P_{\max}
=
|E_1'||E_2'||Y_{12}|
$$

samt

$$
\delta
=
\delta_1-\delta_2
$$

Da kan effektuttrykket skrives

$$
P_e
=
P_c+
P_{\max}
\cos(\delta-\theta_{12})
$$

Dette er den generelle effektvinkelkurven.

---

# Tapsløst nett

For et tapsløst nett gjelder tilnærmet

$$
G_{11}\approx 0
$$

og

$$
\theta_{12}\approx 90^\circ
$$

Dermed fås

$$
P_c=0
$$

og

$$
P_e
=
P_{\max}\sin\delta
$$

som er den klassiske effektvinkelkurven som benyttes i de fleste SMIB-analyser.

---

# Rolle til Kron-reduksjon

Kron-reduksjonen er bindeleddet mellom lastflyt og transientstabilitet.

Studentene kan starte med et ordinært nettverk:

```text
Generator
   |
Transformer
   |
Bus 1
 /   \
Linje Linje
 \   /
Bus 2
   |
Infinite Bus
```

og redusere dette til

```text
E'  ──────  Y12  ──────  V∞
```

slik at hele nettverket beskrives ved noen få admittanser.

---

# Modellering av feil

Ved en feil innføres alltid en eksplisitt feilkoblingsbuss.

Eksempel:

```text
Bus1 ---- Bus3 ---- Bus2
             |
           Fault
```

Dette gjør at samme metode kan brukes for

- håndregning
- Y-buss
- Kron-reduksjon
- PowerFactory

$$

og dermed

P_{e,\text{pre}}(\delta)
$$

---

## Under feil

$$

$$

og dermed

$$

---

$$

$$

$$

# Koblingen til svingligningen

Når effektvinkelkurven er kjent, kan den settes direkte inn i svingligningen.

Rotorvinkel:

$$
\frac{d\delta}{dt}
=
\omega
$$

Rotorhastighet:

$$
\frac{d\omega}{dt}
=
\frac{\omega_s}{2H}
\left(
P_m-P_e
\right)
$$

Her kommer all informasjon fra nettverket inn gjennom

$$
P_e(\delta)
$$

---

# Euler-metoden

Svingligningen kan løses numerisk med Euler-metoden.

Rotorvinkel:

$$
\delta_{k+1}
=
\delta_k+\Delta t\,\omega_k
$$

Rotorhastighet:

$$
\omega_{k+1}
=
\omega_k+
\Delta t
\frac{\omega_s}{2H}
\left(
P_m-P_e(\delta_k)
\right)
$$

---

# Stabilitetsanalyse uten like-areal-kriteriet

I stedet for å beregne kritisk ryddevinkel kan studentene analysere den numeriske løsningen direkte.

Målet er å undersøke om:

$$
\omega \rightarrow 0
$$

og om rotorvinkelen nærmer seg en ny likevekt.

# Kobling mellom SMIB, Euler-metoden og PowerFactory

## Formål

# Faglig progresjon

```text
Lastflyt
   ↓
Intern generatorspenning E'
   ↓
Y-buss
   ↓
Kron-reduksjon
   ↓
To-port-modell
   ↓
Effektvinkelkurve Pe(δ)
   ↓
Svingligning
   ↓
Euler-metoden
   ↓
Transient stabilitet
   ↓
PowerFactory-verifikasjon
```

---

# Kotharis to-port-modell

Etter nettverksreduksjon representeres systemet ved

```text
E'  ────── Stabilitetsnettverk ────── V∞
```

eller matematisk

$$
\mathbf{Y}_{BUS}
=
\begin{bmatrix}
Y_{11} & Y_{12} \\
Y_{21} & Y_{22}
\end{bmatrix}
$$

Aktiv effekt ved generatoren kan uttrykkes som

$$
P_e
=
P_c
+
P_{\max}
\cos(\delta-\theta_{12})
$$

der

$$
P_c
=
|E'|^2G_{11}
$$

og

$$
P_{\max}
=
|E'||V_\infty||Y_{12}|
$$

For et tapsløst nett fås

$$
P_c = 0
$$

og

$$
P_e
=
P_{\max}\sin\delta
$$

Dette er den klassiske effektvinkelkurven som brukes videre i transientstabiliteten.

---

# Svingligningen

Når effektvinkelkurven er bestemt kan den settes inn i svingligningen.

Rotorvinkel:

$$
\frac{d\delta}{dt}
=
\omega
$$

Rotorhastighet:

$$
\frac{d\omega}{dt}
=
\frac{\omega_s}{2H}
\left(
P_m-P_e
\right)
$$

der

- \(P_m\) = mekanisk effekt
- \(P_e\) = elektrisk effekt
- \(H\) = treghetskonstant
- \(\omega_s\) = synkron vinkelhastighet

---

# Fysisk tolkning

Når

$$
P_m>P_e
$$

oppnås

$$
\frac{d\omega}{dt}>0
$$

Generatoren akselererer.

---

Når

$$
P_m<P_e
$$

oppnås

$$
\frac{d\omega}{dt}<0
$$

Generatoren bremses.

---

Definer akselerasjonseffekt

$$
P_a
=
P_m-P_e
$$

som gir:

```text
Pa > 0  → akselerasjon

Pa < 0  → retardasjon

Pa = 0  → likevekt
```

---

# Euler-metoden

Svingligningen løses numerisk ved

$$
\delta_{k+1}
=
\delta_k
+
\Delta t\,\omega_k
$$

og

$$
\omega_{k+1}
=
\omega_k
+
\Delta t
\frac{\omega_s}{2H}
\left(
P_m-P_e(\delta_k)
\right)
$$

Euler-metoden gir dermed rotorvinkel og rotorhastighet som funksjon av tid.

---

# Hvorfor bruke PowerFactory?

PowerFactory kan brukes til å generere referanseforløp for

$$
\delta(t)
$$

og

$$
\omega(t)
$$

etter en feil.

Studentene kan dermed:

1. Bygge modellen.
2. Kjør simulering.
3. Observere plottet.
4. Utføre Euler-beregninger på utvalgte deler av responsen.

---

# Hvor bør Euler-beregningen utføres?

Det er lite pedagogisk verdi i å starte ved

$$
t=0
$$

siden systemet da ligger nær stasjonær drift.

Tilsvarende er det lite interessant å beregne langt ute i et tydelig ustabilt forløp.

Det mest interessante området er rundt det som kan kalles et **knekkpunkt**.

---

# Knekkpunktet

Knekkpunktet er området der systemets videre utvikling avgjøres.

Typisk skjer dette kort tid etter at feilen er fjernet.

```text
δ

|
|                     Ustabil
|                   /
|                  /
|                 /
|                /
|           *
|         *
|       *
|     *
|____*
|
+---------------------------> t

     Knekkpunkt
```

I dette området vet man ennå ikke om systemet vil:

- vende tilbake mot synkron drift,
- eller miste synkronismen.

---

$$

$$

$$

$$

$$

Disse brukes som startverdier i Euler-integrasjonen.

Studentene trenger dermed ikke beregne hele feilforløpet.

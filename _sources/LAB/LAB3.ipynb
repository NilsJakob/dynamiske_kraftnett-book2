# LAB3: Early Stage Assessment – Individuelt prosjektforslag

## Formål

Formålet med oppgaven er å komme tidlig i gang med prosjektarbeidet og motta tilbakemelding på prosjektets faglige relevans, omfang og gjennomførbarhet.

Prosjektforslaget skal fungere som et første utkast til prosjektplan for rapportdelen i ELK330.

---

## Valg av prosjekt

Du kan enten:

- velge en av de anbefalte prosjektretningene:
  - Kortslutningsanalyse og vernkoordinering
  - Integrering av inverterbaserte ressurser (IBR) i distribusjonsnettet

- foreslå et eget prosjekt som er relevant for læringsutbyttene i emnet.

Prosjektet skal inneholde modellering, simulering og analyse av elektriske anlegg eller kraftsystemer.

Ved tvil om prosjektets relevans eller omfang bør faglærer kontaktes tidlig i prosessen.

---

## Innlevering

Oppgaven er individuell.

Følgende skal leveres:

- PDF generert fra LaTeX
- LaTeX-kildekode
- BibTeX-fil
- lenke til Git-repositorium

---

## Omfang

Prosjektforslaget skal normalt være **1-3 side(r)** eksklusive referanser og eventuelle vedlegg.

Målet er å beskrive hva du planlegger å gjøre, ikke å gjennomføre prosjektet.

---

## Innhold

### 1. Prosjekttittel

Gi prosjektet en kort og presis tittel.

---

### 2. Problemstilling og mål

Beskriv kort:

- hva som skal undersøkes
- hvorfor problemstillingen er relevant
- hvilke faglige spørsmål prosjektet skal besvare
- hva du ønsker å oppnå gjennom prosjektet

---

### 3. Faglig grunnlag

Beskriv kort:

- aktuell teori og faglig bakgrunn
- sentrale kilder som er relevante for prosjektet
- hvordan du planlegger å gjennomføre litteratursøket

Det skal benyttes minst:

- én fagbok
- én vitenskapelig artikkel
- én faglig rapport

#### Referanser

Referanser skal håndteres med BibTeX og følge IEEE-stilen.

IEEE-stilen benytter nummererte referanser i teksten:

> Lastflytanalyse er et grunnleggende verktøy innen kraftsystemanalyse [1].

> Integrasjon av distribuert produksjon kan påvirke spenningsprofilen i distribusjonsnettet [2], [3].

Referansene nummereres etter første forekomst i dokumentet og listes samlet i referanselisten.

Eksempler:

[1] O. I. Elgerd, *Electric Energy Systems Theory: An Introduction*. New York, NY, USA: McGraw-Hill, 1982.

[2] P. Kundur, *Power System Stability and Control*. New York, NY, USA: McGraw-Hill, 1994.

[3] IEEE Power & Energy Society, *IEEE Recommended Practice for Industrial and Commercial Power System Analysis*, IEEE Std 3002.2-2018, 2018.

Det anbefales å bruke følgende pakker:

```latex
\usepackage{hyperref}
\usepackage{cite}
```


### 4. Foreløpig plan

Beskriv kort:

- hvordan du ser for deg å angripe prosjektet
- hvilke analyser eller undersøkelser som kan være aktuelle
- forventet arbeidsinnsats fram mot sluttrapporten

Det er ikke nødvendig å beskrive detaljerte modeller eller simuleringer på dette tidspunktet.



### 5. Ambisjon og gjennomførbarhet
 
Reflekter kort over prosjektets planlagte ambisjonsnivå. Beskriv hvordan du vurderer prosjektets omfang og vanskelighetsgrad sett opp mot egne ambisjoner. Diskuter gjerne:
- hvor omfattende modellen eller analysene forventes å bli
- hvilke deler av prosjektet som forventes å være mest utfordrende
- hvilke antakelser eller forenklinger som kan bli nødvendige
- hvordan du har balansert faglige ambisjoner mot et realistisk arbeidsomfang
 
Vurder kort hvilke forutsetninger, parametere eller modellvalg som forventes å ha størst innvirkning på resultatene. Begrunn hvorfor.


## Formelle krav

Besvarelsen skal demonstrere:

- bruk av LaTeX
- figur- og tabellnummerering
- kryssreferanser
- hyperlenker
- BibTeX
- IEEE-referansestil
- korrekt akademisk kildebruk

Dokumentet skal kompilere uten feil som påvirker PDF-dokumentet.

Alle figurer, tabeller, ligninger og referanser skal være korrekt referert fra teksten.

Skriv kort om du foretrekker å lever prosjektet som:

- Pdf i kodet i Latex
- Webside (jupyter book) basert på ferdigprodusert mal

Merk at begge disse innleveringsformene bygger på
- LaTeX-kildekode kombinert med markdown

## Tilbakemelding

Tilbakemeldingen skal gi grunnlag for videre utvikling av prosjektet og den avsluttende prosjektrapporten.

Det vil særlig gis tilbakemelding på:

- prosjektets faglige relevans
- prosjektets omfang og gjennomførbarhet
- problemstilling og målsetting
- litteraturgrunnlag og kildebruk
- den foreløpige planen for videre arbeid

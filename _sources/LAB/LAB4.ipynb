## 5. Tidsplan og estimert arbeidsinnsats

Lag en foreløpig plan for tidsbruk per student.

Bruk følgende mal og juster ved behov:

### Timebruk per student

- Litteratursøk: 20 timer
- Modellering: 40 timer
- Simulering: 30 timer
- Analyse: 25 timer
- Rapportskriving: 35 timer

**Totalt per student: 150 timer**

Dette tilsvarer omtrent 25 timer per uke over en periode på 6 uker.

Beregn også prosjektets totale timeforbruk:

$$
\text{Totalt prosjekt}
=
(\text{timer per student})
\times
(\text{antall studenter})
$$

**Eksempel:**

For en gruppe på 3 studenter:

$$
150 \times 3 = 450 \text{ timer}
$$

For en gruppe på 4 studenter:

$$
150 \times 4 = 600 \text{ timer}
$$

---

**Merk:** Tidsestimatene er veiledende. Gruppen kan omfordele timer mellom aktivitetene dersom prosjektets karakter tilsier det, men den totale arbeidsinnsatsen bør være realistisk i forhold til et prosjekt i et emne på 10 studiepoeng.## 6. Fremdriftsplan

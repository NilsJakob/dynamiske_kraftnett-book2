# Bestemmelse av $\delta_{cr}$ og $t_{cr}$ ved EAC og Stevenson

## Data

Før feil:

$$
P_e = 2.10\sin\delta
$$

Under feil:

$$
P_e = 0.808\sin\delta
$$

Etter feilutkobling:

$$
P_e = 1.50\sin\delta
$$

Mekanisk effekt:

$$
P_m=1.0
$$

# 1. Før-feil driftsvinkel

Før feilen gjelder

$$
P_m=P_{\max,\text{prefault}}\sin\delta_0
$$

slik at

$$
1.0=2.10\sin\delta_0
$$

$$
\delta_0
=
\sin^{-1}\!\left(\frac{1}{2.10}\right)
$$

$$
\delta_0
=
28.44^\circ
$$


# 2. Ustabil post-fault-likevekt

Etter feilutkobling:

$$
P_e=1.50\sin\delta
$$

og

$$
P_m=P_e.
$$

Dermed

$$
1.0=1.50\sin\delta
$$

Den stabile løsningen er

$$
\delta_s
=
\sin^{-1}\!\left(\frac{1}{1.50}\right)
=
41.81^\circ
$$

Den ustabile løsningen er

$$
\delta_u
=
180^\circ-41.81^\circ
$$

$$
\delta_u
=
138.19^\circ
$$



# 3. Likearealkriteriet

Kritisk stabilitet oppstår når

$$
A_1=A_2
$$

der

$$
A_1
=
\int_{\delta_0}^{\delta_{cr}}
(1-0.808\sin\delta)\,d\delta
$$

og

$$
A_2
=
\int_{\delta_{cr}}^{\delta_u}
(1.50\sin\delta-1)\,d\delta.
$$

Etter integrasjon:

$$
P_m(\delta_u-\delta_0)
=
0.808(\cos\delta_0-\cos\delta_{cr})
+
1.50(\cos\delta_{cr}-\cos\delta_u)
$$

Samle leddene med $\cos\delta_{cr}$:

$$
\cos\delta_{cr}
=
\frac{
P_m(\delta_u-\delta_0)
-
0.808\cos\delta_0
+
1.50\cos\delta_u
}
{1.50-0.808}
$$


# 4. Innsetting av tall

Konverter til radianer:

$$
\delta_0
=
28.44^\circ
=
0.4964
$$

$$
\delta_u
=
138.19^\circ
=
2.4120
$$

Videre:

$$
\cos\delta_0
=
0.8793
$$

$$
\cos\delta_u
=
-0.7454
$$

Setter inn:

$$
\cos\delta_{cr}
=
\frac{
(2.4120-0.4964)
-
0.808(0.8793)
+
1.50(-0.7454)
}
{0.692}
$$

$$
\cos\delta_{cr}
=
0.128
$$

$$
\delta_{cr}
=
\cos^{-1}(0.128)
$$

$$
\boxed{\delta_{cr}=82.6^\circ}
$$


# 5. Stevenson-beregning

Gitt

$$
\Delta t=0.05\text{ s}
$$

og

$$
\frac{(\Delta t)^2}{M}
=
8.33
$$

Stevensons formler:

$$
\Delta(\Delta\delta)
=
8.33\Delta P
$$

$$
\Delta\delta_n
=
\Delta\delta_{n-1}
+
8.33\Delta P_{n-1}
$$

$$
\delta_n
=
\delta_{n-1}
+
\Delta\delta_n
$$

### Viktig observasjon

Feilen kobles ut ved

$$
t=0.20\text{ s}
$$

og da endres

$$
P_{\max}
$$

fra

$$
0.808
$$

til

$$
1.50.
$$

Derfor blir

$$
P_e>P_m
$$

og

$$
\Delta P<0.
$$

Vi ser da at

$$
\Delta\delta_4=14.76^\circ
$$

$$
\Delta\delta_5=11.78^\circ
$$

$$
\Delta\delta_6=7.97^\circ
$$

Rotorhastigheten avtar kontinuerlig. Fortsetter beregningen noen få steg til, får vi

$$
\Delta\delta=0
$$

og deretter

$$
\Delta\delta<0,
$$

slik at rotorvinkelen når et maksimum og begynner å avta igjen.

Dermed er systemet transient stabilt.

```python
# TEST CELL
import numpy as np
import matplotlib.pyplot as plt

# ----------------------------------
# Stabilt tilfelle
# Feil utkoblet ved t = 0.20 s
# ----------------------------------

t = np.array([
    0.00,
    0.05,
    0.10,
    0.15,
    0.20,
    0.25,
    0.30,
    0.35,
    0.40
])

delta = np.array([
    28.44,
    31.01,
    38.45,
    50.04,
    64.80,
    76.58,
    84.55,
    88.40,
    88.09
])

Pmax = np.array([
    0.808,
    0.808,
    0.808,
    0.808,
    1.500,
    1.500,
    1.500,
    1.500,
    1.500
])

Pe = np.array([
    0.385,
    0.416,
    0.502,
    0.619,
    1.358,
    1.458,
    1.494,
    1.499,
    1.499
])

DeltaP = np.array([
     0.308,
     0.584,
     0.498,
     0.381,
    -0.358,
    -0.458,
    -0.494,
    -0.499,
    -0.499
])

DeltaDelta = np.array([
     2.57,
     4.87,
     4.15,
     3.17,
    -2.98,
    -3.81,
    -4.12,
    -4.16,
    -4.16
])

Delta_delta = np.array([
     0.00,
     2.57,
     7.44,
    11.59,
    14.76,
    11.78,
     7.97,
     3.85,
    -0.31
])

# Kritiske størrelser
delta_cr = 82.6
t_cr = 0.252
t_clear = 0.20

# Finn maksimum rotorvinkel
imax = np.argmax(delta)

# ----------------------------------
# Plot
# ----------------------------------

fig, ax = plt.subplots(3, 1, figsize=(11, 10), sharex=True)

# --------------------
# Rotorvinkel
# --------------------
ax[0].plot(
    t,
    delta,
    'bo-',
    linewidth=2,
    label=r'$\delta$'
)

ax[0].axvline(
    t_clear,
    color='red',
    linestyle='--',
    linewidth=2,
    label='Feil utkoblet'
)

ax[0].axhline(
    delta_cr,
    color='green',
    linestyle=':',
    linewidth=2,
    label=r'$\delta_{cr}=82.6^\circ$'
)

ax[0].plot(
    t[imax],
    delta[imax],
    'ro',
    markersize=8
)

ax[0].annotate(
    rf'$\delta_{{max}}={delta[imax]:.1f}^\circ$',
    xy=(t[imax], delta[imax]),
    xytext=(0.28, 96),
    arrowprops=dict(arrowstyle='->')
)

ax[0].set_ylabel(r'$\delta$ (grader)')
ax[0].set_title('Stevenson-metoden: Stabilt tilfelle')
ax[0].grid(True)
ax[0].legend()

# --------------------
# Effekter
# --------------------
ax[1].plot(
    t,
    Pe,
    'go-',
    linewidth=2,
    label=r'$P_e$'
)

ax[1].plot(
    t,
    Pmax,
    'k--',
    linewidth=2,
    label=r'$P_{max}$'
)

ax[1].plot(
    t,
    DeltaP,
    'rs-',
    linewidth=2,
    label=r'$\Delta P=P_m-P_e$'
)

ax[1].axvline(
    t_clear,
    color='red',
    linestyle='--'
)

ax[1].axhline(
    0,
    color='black',
    linewidth=1
)

ax[1].set_ylabel('Effekt (p.u.)')
ax[1].grid(True)
ax[1].legend()

# --------------------
# Stevenson Δδ
# --------------------
ax[2].plot(
    t,
    Delta_delta,
    'mo-',
    linewidth=2,
    label=r'$\Delta\delta$'
)

ax[2].axvline(
    t_clear,
    color='red',
    linestyle='--'
)

ax[2].axhline(
    0,
    color='black',
    linewidth=1
)

ax[2].plot(
    t[-1],
    Delta_delta[-1],
    'ro',
    markersize=8
)

ax[2].annotate(
    r'$\Delta\delta \approx 0$',
    xy=(t[-1], Delta_delta[-1]),
    xytext=(0.24, 5),
    arrowprops=dict(arrowstyle='->')
)

ax[2].set_xlabel('Tid (s)')
ax[2].set_ylabel(r'$\Delta\delta$ (grader)')
ax[2].grid(True)
ax[2].legend()

plt.tight_layout()
plt.show()
```

## 1. Spenning bak transientreaktansen

Før feil er linjene parallelle:

$$
X_{\mathrm{nett}} = X_t + \frac{X_L}{2} = 0{,}10 + 0{,}25 = 0{,}35\ \text{p.u.}
$$

Effekt-vinkel-ligningen mellom terminalen og infinitetsbussen gir

$$
\sin\alpha = X_{\mathrm{nett}} = 0{,}35,\qquad \alpha = 20{,}49^\circ.
$$

Dermed er

$$
V_t = 1\angle 20{,}49^\circ \approx 0{,}93675 + j0{,}35000\ \text{p.u.}
$$

Strømmen fra generatoren mot nettet er

$$
I_a = \frac{V_t - V_\infty}{jX_{\mathrm{nett}}} \approx 1{,}000 + j0{,}18072\ \text{p.u.}
$$

Spenningen bak transientreaktansen blir

$$
\begin{aligned}
E' &= V_t + jX'_d I_a \\
&= (0{,}93675 + j0{,}35000) + j0{,}25(1{,}000 + j0{,}18072) \\
&\approx 0{,}89157 + j0{,}60000.
\end{aligned}
$$

Altså

$$
\boxed{E' \approx 1{,}0747\angle 33{,}94^\circ\ \text{p.u.}}
$$

## 2. Største overførbare effekt

For hvert driftstilfelle er $P_e = P_{\max}\sin\delta$.

**Friskt system.** Total overføringsreaktans, $X_{transfer}$ er

$$
X_{transfer,I} = X'_d + X_t + \frac{X_L}{2} = 0,25+0,1+\frac{0,5+0,5}{2}= 0{,}60\ \text{p.u.}
$$

$$
P_{\max I} = \frac{|E'||V_\infty|}{X_{transfer,I}} = \frac{1{,}0747}{0{,}60} \approx \boxed{1{,}791\ \text{p.u.}}
$$

### To port modell
#### Finne ny øverføringsreaktans mellom de to portene port 1 og 2
##### Friskt system – to parallelle linjer

Buss 1 er generatorens indre node, buss 2 er infinitetsbussen,
og buss 3 er knutepunktet etter transformatoren. 

Siden bus 3 ikke er spenningsatt er her $I_3 = 0$ og vi kan bruke kron reduksjon til å eliminere buss 3 og sitte igjen med overføringadmittans mellom buss 1 og 2; $Y'_{12}$

$$
y_{12}=0
\qquad
y_{13}=\frac{1}{j(0{,}25+0{,}10)}
=-j\frac{20}{7},
\qquad
y_{23}=\frac{1}{j0{,}50}+\frac{1}{j0{,}50}

$$

Admittansmatrisen før feil er

$$
Y_{\mathrm{før}}=
\begin{bmatrix}
-j\frac{20}{7} & 0 & j\frac{20}{7}\\
0 & -j4 & j4\\
j\frac{20}{7} & j4 & -j\frac{48}{7}
\end{bmatrix}.
$$

Her er $Y_{22}=-j4$, mens
$Y_{33}=-j\frac{20}{7}-j4=-j\frac{48}{7}$.

Buss $n=3$ elimineres med Kron-reduksjon:

$$
Y_{ki,\mathrm{ny}}
=
Y_{ki,\mathrm{gammel}}
-
\frac{
Y_{kn,\mathrm{gammel}}Y_{ni,\mathrm{gammel}}
}{
Y_{nn,\mathrm{gammel}}
}
$$

Koblingsleddet mellom buss 1 og 2 blir

$$
\begin{aligned}
Y_{12,\mathrm{ny}}
&=Y_{12,\mathrm{gammel}}
-\frac{Y_{13,\mathrm{gammel}}Y_{32,\mathrm{gammel}}}
{Y_{33,\mathrm{gammel}}}\\
&=0-\frac{(j20/7)(j4)}{-j48/7}\\
&=j\frac{5}{3}.
\end{aligned}
$$

Tilsvarende blir diagonalelementene

$$
\begin{aligned}
Y_{11,\mathrm{ny}}
&=-j\frac{20}{7}
-\frac{(j20/7)^2}{-j48/7}
=-j\frac{5}{3},\\
Y_{22,\mathrm{ny}}
&=-j4-\frac{(j4)^2}{-j48/7}
=-j\frac{5}{3}.
\end{aligned}
$$



Den reduserte matrisen, med bussrekkefølge $(1,2)$, er dermed

$$
Y'_{\mathrm{før,red}}=
\begin{bmatrix}
Y'_{11} & Y'_{12}\\
Y'_{21} & Y'_{22}
\end{bmatrix}.
\qquad
Y_{\mathrm{før,red}}=
\begin{bmatrix}
-j\frac{5}{3} & j\frac{5}{3}\\
j\frac{5}{3} & -j\frac{5}{3}
\end{bmatrix}.
$$

### Bevis
Vi får da at kron-reduksjon fungerer fordi vi finner at overføringsadmittansen er det inverse av overføringsreaktansen $$Y'_{12}=\frac{1}{X_{transfer,1}}=\frac{1}{j0,6}=j\frac{5}{3}$$
Fra effektligningen får vi $P_{max,I}$
$$P_{max,I} = |E_1||E_2|Y'_{12}| = 1,0747\cdot1\cdot\frac{5}{3}=1,7911$$

### Trefasekortslutning midt på én linje

Den friske linjen er fortsatt koblet mellom buss 3 og 2. Den feilrammede linjen deles i to halvdeler, hver med reaktans $j0{,}25$ og admittans $-j4$. Feilpunktet $F$ er kortsluttet til jord.

Før feilpunktet jordes, har nettverket fire noder med rekkefølgen $(1,3,2,F)$:

$$
Y_{\mathrm{feil,4}}=
\begin{bmatrix}
-j\frac{20}{7} & j\frac{20}{7} & 0 & 0\\
j\frac{20}{7} & -j\frac{62}{7} & j2 & j4\\
0 & j2 & -j6 & j4\\
0 & j4 & j4 & -j8
\end{bmatrix}.
$$

Siden $V_F=0$, er feilpunktet en **jordet node**, ikke en intern node med ukjent spenning som skal Kron-reduseres. De to halvlinjene bidrar derfor som shuntadmittanser til diagonalen ved buss 3 og buss 2.

![Kraftsystemligning](../images/assignment9b3.png)

Matrisen for de tre gjenværende bussene blir

Med bussrekkefølgen $(1,3,2)$ defineres admittans ut fra generator (mellom buss 1 og 3; $y_e$), admittans på linjestykke (mellom buss 3 og 2; $y_L$) og shuntadmittans (fra feilsted til jord; $y_g$) 

$$
y_e=\frac{1}{j(X'_d+X_t)}
=\frac{1}{j(0.25+0.10)}
=-j\frac{20}{7},
$$


$$
y_L=\frac{1}{j0.50}=-j2,
\qquad
y_f=\frac{1}{j0.25}=-j4,
$$

der $y_L$ er admittansen til den friske linjen, og $y_f$ er admittansen til hver halvdel av den feilrammede linjen mot jord.

Admittansmatrisen under feilen for de tre gjenværende bussene blir
$$
Y_{\mathrm{feil}}=
\begin{bmatrix}
y_e & -y_e & 0\\
-y_e & y_e+y_L+y_g & -y_L\\
0 & -y_L & y_L+y_g
\end{bmatrix}.
$$

Ved innsetting av verdiene fås
$$
\boxed{
Y_{\mathrm{feil}}=
\begin{bmatrix}
-j\frac{20}{7} & j\frac{20}{7} & 0\\
j\frac{20}{7} & -j\frac{62}{7} & j2\\
0 & j2 & -j6
\end{bmatrix}.
}
$$

Nå Kron-reduseres buss $n=3$. Koblingsleddet blir

$$
\begin{aligned}
Y_{12,\mathrm{ny}}
&=Y_{12}-\frac{Y_{13}Y_{32}}{Y_{33}}\\
&=0-\frac{(j20/7)(j2)}{-j62/7}\\
&=j\frac{20}{31}.
\end{aligned}
$$

De to diagonalelementene beregnes på samme måte:

$$
\begin{aligned}
Y_{11,\mathrm{ny}}
&=-j\frac{20}{7}
-\frac{(j20/7)^2}{-j62/7}
=-j\frac{60}{31},\\[4pt]
Y_{22,\mathrm{ny}}
&=-j6
-\frac{(j2)^2}{-j62/7}
=-j\frac{172}{31}.
\end{aligned}
$$

Dermed er

$$
Y_{\mathrm{feil,red}}=
\begin{bmatrix}
-j\frac{60}{31} & j\frac{20}{31}\\
j\frac{20}{31} & -j\frac{172}{31}
\end{bmatrix}.
$$

Diagonalelementene er ikke motsatt lik koblingsleddet fordi nettet under feilen også har forbindelser til jord. Det er koblingsleddet $Y_{12,\mathrm{ny}}=j20/31$ som bestemmer koeffisienten foran $\sin\delta$:

$$
\boxed{
P_{\max II}=|E'||V_\infty|\frac{20}{31}
\approx0{,}693\ \text{p.u.}
}
$$

## 3. Effekt–vinkel-kurvene

$$
\begin{aligned}
P_{e,I}(\delta) &= 1{,}791\sin\delta,\\
P_{e,II}(\delta) &= 0{,}693\sin\delta,\\
P_{e,III}(\delta) &= 1{,}264\sin\delta.
\end{aligned}
$$

Tegn alle tre for $0^\circ \le \delta \le 180^\circ$. Alle er null ved $0^\circ$ og $180^\circ$, og har topp ved $90^\circ$. Kurven for friskt system ligger høyest, kurven under feil lavest, og kurven med én linje frakoblet ligger mellom dem. Driftspunktet før feil er $\delta_0 \approx 33{,}94^\circ$ ved $P_m = 1{,}0$ p.u.

```python
# TEST CELL
import numpy as np
import matplotlib.pyplot as plt

# Vinkelområde i grader
delta_deg = np.linspace(0, 180, 500)
delta_rad = np.deg2rad(delta_deg)

# Maksimal effekt for hvert driftstilfelle [p.u.]
Pmax_I = 1.791
Pmax_II = 0.693
Pmax_III = 1.264
Pm = 1.0

# Effekt-vinkel-kurver
Pe_I = Pmax_I * np.sin(delta_rad)
Pe_II = Pmax_II * np.sin(delta_rad)
Pe_III = Pmax_III * np.sin(delta_rad)

# Plot
plt.figure(figsize=(9, 5.5))
plt.plot(delta_deg, Pe_I, label=r"Før feil: $P_{\max I}=1.791$ p.u.")
plt.plot(delta_deg, Pe_II, label=r"Under feil: $P_{\max II}=0.693$ p.u.")
plt.plot(delta_deg, Pe_III, label=r"Etter feil: $P_{\max III}=1.264$ p.u.")
plt.axhline(Pm, color="black", linestyle="--", label=r"$P_m=1.0$ p.u.")

plt.xlim(0, 180)
plt.ylim(0, 2.0)
plt.xticks(np.arange(0, 181, 30))
plt.xlabel(r"Rotorvinkel $\delta$ [grader]")
plt.ylabel(r"Elektrisk effekt $P_e$ [p.u.]")
plt.title("Effekt–vinkel-kurver for SMIB-systemet")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

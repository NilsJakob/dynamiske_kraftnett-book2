# Øving 9: Kritisk klareringstid (Stevenson-type med to-port-data)

## Løsningsforslag

### Gitte data

Generatorspenning bak transient reaktans:

$$
E' = 1.15\ pu
$$

Uendelig sterk skinne:

$$
V = 1.0\ pu
$$

Generatorens transiente reaktans:

$$
X_d' = 0.25\ pu
$$

Linjereaktans:

$$
X_L = 0.50\ pu
$$

Mekanisk effekt:

$$
P_m = 0.90\ pu
$$

Treghetskonstant:

$$
H = 5.0\ \text{MJ/MVA}
$$

Frekvens:

$$
f = 50\ \text{Hz}
$$

Under feilen:

$$
P_e=0
$$

---

## a) Total overføringsreaktans

Den totale reaktansen mellom generatorens indre spenning og nettet er

$$
X=X_d'+X_L
$$

Innsatt tallverdier:

$$
X=0.25+0.50
$$

$$
\boxed{X=0.75\ pu}
$$

---

## b) Maksimal overførbar effekt

For den klassiske generator-modellen er effekt-vinkel-karakteristikken

$$
P_e=P_{\max}\sin\delta
$$

der

$$
P_{\max}=\frac{E'V}{X}
$$

Innsatt tall:

$$
P_{\max}
=
\frac{1.15\cdot1.0}{0.75}
$$

$$
P_{\max}=1.533
$$

Dermed

$$
\boxed{P_{\max}=1.53\ pu}
$$

---

## c) Initial rotorvinkel

Før feilen er systemet i stasjonær drift, slik at

$$
P_m=P_e
$$

Dermed

$$
P_m=P_{\max}\sin\delta_0
$$

og

$$
\delta_0
=
\sin^{-1}
\left(
\frac{P_m}{P_{\max}}
\right)
$$

Innsatt verdier:

$$
\delta_0
=
\sin^{-1}
\left(
\frac{0.90}{1.533}
\right)
$$

$$
\delta_0
=
\sin^{-1}(0.587)
$$

$$
\delta_0
=
35.96^\circ
$$

eller i radianer

$$
\delta_0
=
35.96\frac{\pi}{180}
=
0.6276\ \text{rad}
$$

Altså

$$
\boxed{\delta_0=35.96^\circ}
$$

---

## d) Kritisk clearingvinkel

For en trefase terminalfeil hvor

$$
P_e=0
$$

kan Stevenson-formelen brukes direkte:

$$
\delta_{cr}
=
\cos^{-1}
\left[
(\pi-2\delta_0)\sin\delta_0
-\cos\delta_0
\right]
$$

Vi bruker

$$
\delta_0=0.6276\ \text{rad}
$$

Først:

$$
\sin\delta_0=\sin(0.6276)=0.587
$$

$$
\cos\delta_0=\cos(0.6276)=0.809
$$

Videre:

$$
\pi-2\delta_0
=
3.1416-2(0.6276)
=
1.8864
$$

Dermed:

$$
(\pi-2\delta_0)\sin\delta_0
=
1.8864\cdot0.587
=
1.107
$$

og

$$
1.107-0.809
=
0.298
$$

Så:

$$
\delta_{cr}
=
\cos^{-1}(0.298)
$$

$$
\delta_{cr}
=
1.268\ \text{rad}
$$

eller

$$
\delta_{cr}
=
72.7^\circ
$$

Dermed

$$
\boxed{\delta_{cr}=72.7^\circ}
$$

---

## e) Synkron vinkelhastighet

Synkron vinkelhastighet er

$$
\omega_s
=
2\pi f
$$

Innsatt:

$$
\omega_s
=
2\pi\cdot50
$$

$$
\omega_s
=
314.16\ \text{rad/s}
$$

Altså

$$
\boxed{\omega_s=314.16\ \text{rad/s}}
$$

---

## f) Rotorakselerasjon under feilen

Under feilen er

$$
P_e=0
$$

slik at akselerasjonseffekten blir

$$
P_a=P_m-P_e
$$

$$
P_a=0.90
$$

Svingeligningen gir

$$
\alpha
=
\frac{\omega_s}{2H}P_a
$$

eller

$$
\alpha
=
\frac{314.16}{2\cdot5}
\cdot0.90
$$

$$
\alpha
=
31.416\cdot0.90
$$

$$
\alpha
=
28.27\ \text{rad/s}^2
$$

Dermed

$$
\boxed{\alpha=28.27\ \text{rad/s}^2}
$$

---

## g) Kritisk clearingtid

Formelen er

$$
t_{cr}
=
\sqrt{
\frac{2(\delta_{cr}-\delta_0)}
{\alpha}
}
$$

Vi bruker vinklene i radianer:

$$
\delta_{cr}=1.268
$$

$$
\delta_0=0.6276
$$

Differansen blir

$$
\delta_{cr}-\delta_0
=
1.268-0.6276
=
0.6404
$$

Dermed

$$
t_{cr}
=
\sqrt{
\frac{2(0.6404)}
{28.27}
}
$$

$$
t_{cr}
=
\sqrt{0.0453}
$$

$$
t_{cr}
=
0.213\ s
$$

Altså

$$
\boxed{t_{cr}=0.21\ s}
$$

Resultatene stemmer godt med de oppgitte kontrollverdiene:

$$
P_{\max}\approx1.53\ pu
$$

$$
\delta_0\approx36^\circ
$$

$$
\delta_{cr}\approx73^\circ
$$

$$
t_{cr}\approx0.21\ s
$$

$$
\boxed{0.21\ \text{s}}
$$

for å opprettholde transient stabilitet.

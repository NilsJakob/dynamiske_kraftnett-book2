# Øving 7 – Beregning av elektromekaniske konstanter

## Oppgave

Betrakt en **4-polet, 50 Hz, 11 kV turbogenerator**.

Generatoren har følgende merkedata:

$$
S_n = 52.0\ \text{MVA}
$$

$$
\cos\varphi = 0.85
$$

Generatoren har en sylindrisk rotor med treghetsmoment

$$
I = 6500\ \text{kg}\cdot\text{m}^2
$$

### a)

Beregn den mekaniske rotasjonshastigheten

$$
n_{mec}
$$

i rpm.

---

### b)

Beregn den synkrone vinkelhastigheten

$$
\omega_s
$$

i rad/s.

---

### c)

Beregn den kinetiske energien lagret ved synkron hastighet.

$$
KE=\frac{1}{2}I\omega_s^2
$$

Svar i MJ.

---

### d)

Beregn treghetskonstanten

$$
H
$$

i MJ/MVA.

---

### e)

Beregn treghetskonstanten

$$
H
$$

i MJ/MW.

---

### f)

Beregn momentumkonstanten

$$
M
$$

i MJ·s per elektrisk grad.

Bruk

$$
M=\frac{S_nH}{180f}
$$

---

# Svar

## a) Mekanisk rotasjonshastighet

**Svar**

$$
\boxed{n_{mec}=1500.0\ \text{rpm}}
$$

---

## b) Synkron vinkelhastighet


**Svar**

$$
\boxed{\omega_s=314.1593\ \text{rad/s}}
$$

---

## c) Kinetisk energi

**Svar**

$$
\boxed{KE=320.31\ \text{MJ}}
$$

---

## d) Treghetskonstant H i MJ/MVA


**Svar**

$$
\boxed{H=6.16\ \text{MJ/MVA}}
$$

---

## e) Treghetskonstant H i MJ/MW


**Svar**

$$
\boxed{H=7.25\ \text{MJ/MW}}
$$

---

## f) Momentumkonstant M


**Svar**

$$
\boxed{M=0.0356\ \text{MJ}\cdot\text{s/el.grad}}
$$

# Øving 7 – Beregning av rotasjonsinerti for synkrongenerator

## Oppgave

Betrakt en synkrongenerator (SG) tilknyttet et kraftsystem med følgende merkedata:

- $S_n = 250$ MVA
- Effektfaktor:$PF_{rated}=0.90$

- Merkespenning:$V_{rated}=18.0\ \text{kV}$

- Rotasjonshastighet:$n=3000\ \text{rpm}$

- Treghetskonstant: $H=4.8\ \text{MJ/MVA}$

- Systembase: $S_{base}=600\ \text{MVA}$

---

## Oppgaver

### a)

Beregn generatorens aktive merkeeffekt

$$
P_n=S_n \cdot PF_{rated}
$$

### b)

Beregn den synkrone vinkelhastigheten

$$
\omega_s=\frac{2\pi n}{60}
$$

### c)

Beregn den kinetiske energien lagret ved merkedrift

$$
KE=H \cdot S_n
$$

### d)

Bestem generatorens rotasjonsinerti (treghetsmoment)

$$
J=\frac{2KE}{\omega_s^2}
$$

### e)

Beregn treghetskonstanten på systembasen

$$
H_{sys}=H\frac{S_n}{S_{base}}
$$

### f)

Forklar hva som skjer med frekvensresponsen dersom treghetskonstanten $H$ reduseres.

---

# Fasit

## a) Aktiv merkeeffekt

$$
P_n=250\cdot0.90
$$

$$
\boxed{P_n=225\ \text{MW}}
$$

## b) Synkron vinkelhastighet

$$
\omega_s=\frac{2\pi\cdot3000}{60}=314.16\ \text{rad/s}
$$

$$
\boxed{\omega_s=314.16\ \text{rad/s}}
$$

## c) Kinetisk energi

$$
KE=4.8\cdot250=1200\ \text{MJ}
$$

$$
\boxed{KE=1200\ \text{MJ}}
$$

## d) Rotasjonsinerti

$$
J=\frac{2\cdot1200\times10^6}{(314.16)^2}
$$

$$
J\approx24318\ \text{kgm}^2
$$

$$
\boxed{J\approx2.43\times10^4\ \text{kgm}^2}
$$

## e) Treghetskonstant på systembasen

$$
H_{sys}=4.8\cdot\frac{250}{600}=2.0\ \text{MJ/MVA}
$$

$$
\boxed{H_{sys}=2.0\ \text{MJ/MVA}}
$$

# Øving 9 - Løsningsforslag

En generator som opererer ved 50 Hz leverer en mekanisk effekt på

$$
P_m = 0.9 \text{ p.u.}
$$

til en uendelig sterk skinne gjennom et nett der resistansen kan neglisjeres.

Før feilen er den maksimalt overførbare effekten

$$
P_{\max I}=1.6 \text{ p.u.}
$$

Under feilen reduseres den maksimalt overførbare effekten til

$$
P_{\max II}=0.3 \text{ p.u.}
$$

Etter feilutkobling blir den maksimalt overførbare effekten

$$
P_{\max III}=1.2 \text{ p.u.}
$$

Generatorens treghetskonstant er

$$
H=3.0 \text{ p.u.}
$$

og feilen kobles ut etter

$$
t_c=0.15\text{ s}
$$

---

# Del 1: Kritisk utkoblingsvinkel

## Startvinkel

$$
\delta_0
=
\sin^{-1}
\left(
\frac{P_m}{P_{\max I}}
\right)
=
\sin^{-1}
\left(
\frac{0.9}{1.6}
\right)
=
34.23^\circ
$$

---

## Maksimal rotorvinkel

$$
\delta_{\max}
=
180^\circ
-
\sin^{-1}
\left(
\frac{P_m}{P_{\max III}}
\right)
$$

$$
\delta_{\max}
=
180^\circ
-
\sin^{-1}
\left(
\frac{0.9}{1.2}
\right)
=
131.41^\circ
$$

---

## Konvertering til radianer

$$
\delta_0
=
34.23^\circ
=
0.597\text{ rad}
$$

$$
\delta_{\max}
=
131.41^\circ
=
2.293\text{ rad}
$$

---

## Beregning av cosinus til kritisk utkoblingsvinkel

Stevensons ligning:

$$
\cos \delta_{cr}
=
\frac{
P_m(\delta_{\max}-\delta_0)
-
P_{\max II}\cos\delta_0
+
P_{\max III}\cos\delta_{\max}
}
{
P_{\max III}-P_{\max II}
}
$$

Setter inn tall:

$$
\cos \delta_{cr}
=
\frac{
0.9(2.293-0.597)
-
0.3\cos 34.23^\circ
+
1.2\cos 131.41^\circ
}
{
1.2-0.3
}
$$

$$
=
\frac{
1.526
-
0.248
-
0.793
}
{
0.9
}
$$

$$
=
0.539
$$

Dermed blir

$$
\delta_{cr}
=
\cos^{-1}(0.539)
=
57.4^\circ
$$

---

## Resultat

$$
\boxed{\delta_{cr}=57.4^\circ}
$$

---

# Del 2: Svingskurve

## Beregning av M

$$
M
=
\frac{H}{180f}
=
\frac{3.0}{180\cdot50}
=
3.33\times10^{-4}
$$

Tidssteget er

$$
\Delta t=0.05\text{ s}
$$

Dermed

$$
\frac{(\Delta t)^2}{M}
=
\frac{0.05^2}{3.33\times10^{-4}}
=
7.50
$$

---

## Startverdi

$$
\delta_0=34.23^\circ
$$

Like før feilen:

$$
\Delta P = 0
$$

Umiddelbart etter feilen:

$$
\Delta P
=
0.9-0.3\sin34.23^\circ
=
0.731
$$

Første halvtrinn:

$$
\Delta P
=
\frac{0.731}{2}
=
0.366
$$

---

## Steg 1

$$
t=0.05\text{ s}
$$

$$
\Delta\delta
=
0+7.50(0.366)
=
2.74^\circ
$$

$$
\delta
=
34.23+2.74
=
36.97^\circ
$$

---

## Steg 2

$$

$$

$$

$$
\delta

=

$$

---

Siden

$$
\delta(0.15)=58.40^\circ
>
\delta_{cr}=57.4^\circ
$$

viser beregningen at en utkoblingstid på

$$
t_c=0.15\text{ s}
$$

er **litt større enn den kritiske clearingtiden**, og systemet ligger derfor svært nær grensen for transient stabilitet.

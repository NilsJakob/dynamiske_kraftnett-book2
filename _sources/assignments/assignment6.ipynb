## Del 1: Lag en Gauss-modell for solinnstråling

Vi ønsker å beskrive solinnstrålingen gjennom et døgn ved hjelp av en Gauss-kurve:

$$
G(t)=A\exp\left(-\frac{(t-\mu)^2}{2\sigma^2}\right)
$$


### Oppgave 1

Lag et Python-program som:

1. Oppretter en tidsakse fra kl. 00 til kl. 24.
2. Setter maksimal innstråling til 800 W/m².
3. Antar at maksimal innstråling skjer kl. 13.
4. Velger en passende verdi for \(\sigma\).
5. Plotter Gauss-kurven.

#### Eksempel

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 24, 500)

A = 800
mu = 13
sigma = 3

G = A * np.exp(-(t - mu)**2 / (2 * sigma**2))

plt.plot(t, G)
plt.xlabel("Tid [timer]")
plt.ylabel("Innstråling [W/m²]")
plt.grid()
plt.show()
```

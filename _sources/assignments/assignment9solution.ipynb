## 1. Likearealkriteriet

Startvinkelen før feilen er

$$
\delta_0
=\sin^{-1}\left(\frac{P_m}{P_{\max I}}\right)
=\sin^{-1}\left(\frac{0{,}9}{1{,}6}\right)
\approx\boxed{34{,}23^\circ}.
$$

Den maksimale rotorvinkelen etter feilutkobling er

$$
\begin{aligned}
\delta_{\max}
&=180^\circ-\sin^{-1}\left(\frac{P_m}{P_{\max III}}\right)\\
&=180^\circ-\sin^{-1}\left(\frac{0{,}9}{1{,}2}\right)\\
&\approx\boxed{131{,}41^\circ}.
\end{aligned}
$$

Vinkeldifferansen i telleren må settes inn i radianer:

$$
\delta_{\max}-\delta_0
=(131{,}41^\circ-34{,}23^\circ)\frac{\pi}{180}
\approx1{,}6961\ \text{rad}.
$$

Lik-arealkriteriet gir

$$
\begin{aligned}
\cos\delta_{\mathrm{cr}}
&=
\frac{
P_m(\delta_{\max}-\delta_0)
-P_{\max II}\cos\delta_0
+P_{\max III}\cos\delta_{\max}
}{
P_{\max III}-P_{\max II}
}\\
&=
\frac{
0{,}9(1{,}6961)
-0{,}3\cos(34{,}23^\circ)
+1{,}2\cos(131{,}41^\circ)
}{
1{,}2-0{,}3
}\\
&\approx0{,}5386.
\end{aligned}
$$

Dermed blir den kritiske utkoblingsvinkelen

$$
\boxed{
\delta_{\mathrm{cr}}
=\cos^{-1}(0{,}5386)
\approx57{,}42^\circ.
}
$$

## Modifisert Eulers leapfrog metode

Data:

$$
\delta_0=34{,}23^\circ
$$

$$
P_m=0{,}9,\qquad P_{e}=0{,}3\sin\delta
$$

$$
M=\frac{H}{180f}
=\frac{3}{180\cdot50}
=\frac{1}{3000}
$$

$$
\Delta t=0{,}05\ \text{s}
$$

$$
\frac{\Delta t^2}{M}
=\frac{0{,}05^2}{1/3000}
=7{,}5
$$

### Steg 0 ($t=0{,}00$ s)

$$
P_{e,0}=0{,}3\sin(34{,}23^\circ)
\approx0{,}169
$$

$$
\Delta P_0=0{,}9-0{,}169
\approx0{,}731
$$

Ved oppstart brukes middelverdien mellom null og den første akselererende effekten:

$$
\overline{\Delta P}_0
=\frac{0+0{,}731}{2}
\approx0{,}366
$$

$$
\delta_0=34{,}23^\circ
$$

### Steg 1 ($t=0{,}05$ s)

$$
\Delta(\Delta\delta)_1
=7{,}5(0{,}366)
\approx2{,}74^\circ
$$

$$
\Delta\delta_1=0+2{,}74
=2{,}74^\circ
$$

$$
\delta_1=34{,}23+2{,}74
\approx36{,}97^\circ
$$

### Oppsummering

$$
\begin{array}{c|c|c}
\text{Steg} & t\;(\text{s}) & \delta\;(^\circ)\\
\hline
0 & 0{,}00 & 34{,}23\\
1 & 0{,}05 & 36{,}97\\

Den kritiske utkoblingsvinkelen er

$$
\delta_{\mathrm{cr}}\approx57{,}42^\circ.
$$

Den ligger mellom $\delta_2$ og $\delta_3$. Lineær interpolasjon gir

$$
\begin{aligned}
t_{\mathrm{cr}}
&=0{,}10+
\frac{57{,}42-45{,}11}{58{,}41-45{,}11}(0{,}05)\\
&\approx0{,}146\ \text{s}.
\end{aligned}
$$

$$
\boxed{t_{\mathrm{cr}}\approx0{,}15\ \text{s}}
$$

Utkoblingstiden $t_c=0{,}15$ s er litt større enn det interpolerte estimatet $t_{\mathrm{cr}}\approx0{,}146$ s.

Ved utkoblingstidspunktet $t_c=0{,}15$ s er rotorvinkelen omtrent

$$
\delta(t_c)\approx66{,}32^\circ.
$$

Siden

$$
\delta(t_c)>\delta_{\mathrm{cr}}
\qquad
(66{,}32^\circ>57{,}42^\circ),
$$

er feilen koblet ut for sent til at systemet er transient stabilt etter denne beregningen. Etter feilutkobling fortsetter rotorvinkelen å øke og passerer $\delta_{\max}\approx131{,}41^\circ$.

En lineær interpolasjon mellom $t=0{,}10$ s og $t=0{,}15$ s gir et estimat for kritisk utkoblingstid:

$$
\begin{aligned}
t_{\mathrm{cr}}
&\approx0{,}10+
\frac{57{,}42-50{,}51}{66{,}32-50{,}51}(0{,}05)\\
&\approx\boxed{0{,}122\ \text{s}}.
\end{aligned}
$$

Dermed er den oppgitte utkoblingstiden $0{,}15$ s større enn estimert kritisk utkoblingstid.

```python
# TEST CELL
import numpy as np
import matplotlib.pyplot as plt

# Øving 9-data
Pm = 0.9
#Pmax_I = 1.6       # før feil
Pmax_II = 0.3      # under feil
Pmax_III = 1.2     # etter feil
H = 3.0            # s
f = 50.0           # Hz
tc = 0.15          # s

# Svingeligning for rotorvinkel i grader
M = H / (180 * f)
delta0 = np.degrees(np.arcsin(Pm / Pmax_I))

dt = 0.001
t_end = 1.2
t = np.arange(0, t_end + dt, dt)


def simulate(pmax_function):
    """Integrer svingeligningen med velocity-Verlet."""
    delta = np.zeros_like(t)  # grader
    omega = np.zeros_like(t)  # grader/s
    delta[0] = delta0

    def acceleration(delta_deg, time):
        pmax = pmax_function(time)
        return (Pm - pmax * np.sin(np.deg2rad(delta_deg))) / M

    for n in range(len(t) - 1):
        a_n = acceleration(delta[n], t[n])

        delta[n + 1] = (
            delta[n] + omega[n] * dt + 0.5 * a_n * dt**2
        )

        a_next = acceleration(delta[n + 1], t[n + 1])
        omega[n + 1] = omega[n] + 0.5 * (a_n + a_next) * dt

    return delta


# Tre svingeligninger, hver med konstant Pmax
#delta_I = simulate(lambda time: Pmax_I)
delta_II = simulate(lambda time: Pmax_II)
delta_III = simulate(lambda time: Pmax_III)

# Faktisk forløp: feil fram til tc, deretter feilutkoblet nett
delta_piecewise = simulate(
    lambda time: Pmax_II if time < tc else Pmax_III
)

plt.figure(figsize=(10, 6))
#plt.plot(t, delta_I, label=r"Før feil: $P_{\max I}=1.6$ p.u.")
plt.plot(t, delta_II, label=r"Under feil: $P_{\max II}=0.3$ p.u.")
plt.plot(t, delta_III, label=r"Etter feil: $P_{\max III}=1.2$ p.u.")
plt.plot(
    t,
    delta_piecewise,
    "k--",
    linewidth=2,
    label=rf"Feil til $t_c={tc}$ s, deretter etter feil"
)

plt.axvline(tc, color="gray", linestyle=":", label=r"Feilutkobling $t_c$")
plt.axhline(180, color="red", linestyle=":", label=r"$180^\circ$")
plt.xlabel(r"Time $t$ [s]")
plt.ylabel(r"Rotorvinkel $\delta$ [grader]")
plt.title("Svingkurver for de tre driftstilfellene")
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

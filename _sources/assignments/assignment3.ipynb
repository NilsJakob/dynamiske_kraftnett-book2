## Øving 3: Analyser, lagre og versjonskontroller data

### Læringsmål

Etter denne øvingen skal du kunne:

- lese data fra en CSV-fil,
- utføre en enkel analyse i Python,
- lagre resultater til en ny fil,
- bruke Git til å spore endringer,
- publisere resultater til eget repo på GitHub.

Dere har allerede klonet power-system-data i forrige øving.
- Opprett et nytt, tomt GitHub-repo på egen konto med samme navn
- Deretter går dere inn i den eksisterende power-system-data-mappen
- bytter origin til deres eget repo

Arbeidsflyten blir slik:

```bash
git remote remove origin
git remote add origin https://github.com/BRUKERNAVN/power-system-data.git
git push -u origin main
```

Etter dette fungerer vanlig arbeidsflyt:
```bash
git add .
git commit -m "Løst oppgave"
git push
```

#### Bytt arbeidskatalog
```bash
cd power-system-data
```

#### origin 
`origin` girnavnet på GitHub-repoet man jobber mot.

og fås ved kommandoen:
```bash
git remote -v
```

Resultatet viser hva origin faktisk peker til.
Hvis du stå i det repoet som læreren (altså jeg, Nils Jakob) lagde må du fjerne deg fra det repoet og koble til versjonen på ditt eget repo, for å kunne publisere (pushe) til eget repo:

```bash
# Fjern koblingen til lærerrepoet
git remote remove origin

# Koble til eget repo
git remote add origin https://github.com/BRUKERNAVN/power-system-data.git

# Publiser
git push -u origin main
```

## Øving: Analyser, lagre og versjonskontroller data

### Læringsmål

Etter denne øvingen skal du kunne:

- lese data fra en CSV-fil,
- utføre en enkel analyse i Python,
- lagre resultater til en ny fil,
- bruke Git til å spore endringer,
- publisere resultater til GitHub.

---

### Bakgrunn

Repoet inneholder en tidsserie med forbruksdata i Norge:

```text
load/forbruk_2025.csv
```

Datasettet inneholder målinger for hele året 2025.

Målet er å analysere datasettet og lagre resultatene i et eget område av repositoryet.

---

### Oppgave 1: Les inn datasettet

Les inn filen

```text
load/forbruk_2025.csv
```

ved hjelp av Pandas.

Kontroller at:

- dataene er lest korrekt,
- de første radene kan vises,
- tidsstemplene tolkes som dato og klokkeslett.

---

### Oppgave 2: Beregn månedlig gjennomsnitt

Beregn gjennomsnittlig belastning for hver måned.

Resultatet skal inneholde én verdi per måned.

Eksempel:

| Måned | Gjennomsnittlig last (MW) |
|---------|---------|
| Januar | ... |
| Februar | ... |
| Mars | ... |
| ... | ... |

---

### Oppgave 3: Lagre resultatet

Opprett en mappe:

```text
results/
```

Lagre resultatene som:

```text
results/manedlig_last_2025.csv
```

---

### Oppgave 4: Visualiser resultatet

Lag en figur som viser gjennomsnittlig last for hver måned.

Figuren skal inneholde:

- tittel,
- aksetitler,
- rutenett.

Lagre figuren som:

```text
results/manedlig_last_2025.png
```

---

### Oppgave 5: Bruk Git

Undersøk hvilke filer som er endret:

```bash
git status
```

Legg til de nye filene:

```bash
git add results/manedlig_last_2025.csv
git add results/manedlig_last_2025.png
```

Opprett en commit:

```bash
git commit -m "Legg til analyse av månedlig last for 2025"
```

Last opp endringene til GitHub:

```bash
git push
```

Beregn også:

Etter at øvingen er gjennomført skal repositoriet inneholde:

```text
load/
└── forbruk_2025.csv

results/
├── manedlig_last_2025.csv
├── manedlig_last_2025.png
└── manedlig_last_statistikk_2025.csv
```

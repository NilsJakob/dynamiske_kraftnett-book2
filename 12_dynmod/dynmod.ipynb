### Admittansmetoden
Admittansmetoden (ofte omtalt som den modifiserte admittansmetoden eller [Nodal Admittance Matrix method](https://en.wikipedia.org/wiki/Nodal_admittance_matrix) er en systematisk matematisk metode som brukes i elektroteknikk for å analysere og beregne spenninger og strømmer i komplekse elektriske nettverk.

$$I_k = \sum_{i=1}^n Yki\cdot Vi$$

# Fra kraftflyt til kraftvinkelligningen (SMIB)

## 1. Utgangspunkt: Kraftflytligningen

Vi starter med den komplekse effekten ved buss $k$:

$$
\bar S_k^{*}
=
P_k-jQ_k
=
\sum_{i=1}^{n}
\bar Y_{ki}\,\bar V_i\,\bar V_k^{*}
$$

Skriver vi alle størrelser på polar form:

$$
\bar Y_{ki}
=
|Y_{ki}|e^{j\theta_{ki}}
$$

$$
\bar V_i
=
|V_i|e^{j\delta_i}
$$

$$
\bar V_k^{*}
=
|V_k|e^{-j\delta_k}
$$



får vi

$$
P_k-jQ_k
=
\sum_{i=1}^{n}
|Y_{ki}||V_i||V_k|
e^{j(\theta_{ki}+\delta_i-\delta_k)}
$$

Den reelle delen gir aktiv effekt:

$$
P_k
=
\sum_{i=1}^{n}
|V_i||V_k||Y_{ki}|
\cos(\theta_{ki}+\delta_i-\delta_k)
$$

Dette er den vanlige kraftflytligningen.

![Kraftsystemligning](../images/power_flow_eq.png)



# 2. Stabilitetsmodell

I transient stabilitet flyttes bussene fra terminalspenningene til de transiente interne EMF-ene.

Vi erstatter derfor

$$
V_i \rightarrow E'_i
$$

slik at

$$
P_k
=
\sum_{i=1}^{n}
|E'_i||E'_k||Y_{ki}|
\cos(\theta_{ki}+\delta_i-\delta_k)
$$

hvor

- $E'_i$ er transient internspenning
- $\delta_i$ er rotorvinkelen

# 4. Tapsfritt nett

For klassisk transient stabilitet antar vi $R \approx 0$ slik at nettet består av rene reaktanser. Transferreaktansen er $jX$ og dermed:

$$
Y_{21}
=
-\frac{1}{jX}
=
\frac{j}{X}
$$

slik at $|Y_{21}|=\frac{1}{X}$ og $\theta_{21}=90^\circ$

Videre er $\theta_{11}=-90^\circ$

og siden $\cos(-90^\circ)=0$ forsvinner diagonal-leddet.

Da står vi igjen med

$$
P_1
=
\frac{|E'_1||E'_2|}{X}
\cos(90^\circ+\delta_2-\delta_1)
$$

# 5. Fra cosinus til sinus

Bruk identiteten

$$
\cos(90^\circ-x)=\sin(x)
$$

med

$$
x=\delta_1-\delta_2
$$

Da får vi

$$
P_1
=
\frac{|E'_1||E'_2|}{X}
\sin(\delta_1-\delta_2)
$$


# 5. Fra cosinus til sinus

Bruk identiteten

$$
\cos(90^\circ-x)=\sin(x)
$$

med

$$
x=\delta_1-\delta_2
$$

Da får vi

$$
P_1
=
\frac{|E'_1||E'_2|}{X}
\sin(\delta_1-\delta_2)
$$


# 6. Definer kraftvinkelen

Definer den relative rotorvinkelen

$$
\delta
=
\delta_1-\delta_2
$$

Da blir uttrykket

$$
P_1
=
\frac{|E'_1||E'_2|}{X}
\sin\delta
$$

Definer videre

$$
P_{\max}
=
\frac{|E'_1||E'_2|}{X}
$$

slik at

$$
P=P_{\max}\sin\delta
$$

Dette er den generelle kraftvinkelligningen.



# 7. Overgang til SMIB

En infinite bus modelleres som

$$
E'_2=V
$$

med konstant vinkel

$$
\delta_2=0
$$

Systemet blir da

```text
E'∠δ  ---- jXtransfer ----  V∠0
```

slik at

$$
\delta
=
\delta_1
$$

og kraftvinkelligningen reduseres til

$$
P_e
=
\frac{E'V}{X_{\text{transfer}}}
\sin\delta
$$

hvor

$$
X_{\text{transfer}}
=
X_d' + X_{\text{nett}}
$$

---

# Fysisk tolkning

Kraftflyt:

```text
Vt∠α ---- jXnett ---- V∠0
```

$$
P
=
\frac{V_tV}{X_{\text{nett}}}
\sin\alpha
$$

Stabilitet:

```text
E'∠δ ---- j(Xd'+Xnett) ---- V∠0
```

$$
P_e
=
\frac{E'V}
     {X_d'+X_{\text{nett}}}
\sin\delta
$$

Dermed kan kraftvinkelligningen ses som et spesialtilfelle av den ordinære kraftflytligningen når:

1. Bussene flyttes til de interne EMF-punktene $E'$.
2. Nettet er tapsfritt.
3. Systemet reduseres til to porter.
4. Den ene porten gjøres til en infinite bus.

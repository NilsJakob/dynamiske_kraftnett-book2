# Finne kraftvinkel

### Bestemmelse av effekt-vinkel-ligningen

En synkrongenerator er koblet til et stort kraftsystem representert ved en uendelig buss gjennom parallelle transmisjonslinjer, modellert som SMIB-modell. Gitte driftsbetingelser er: generatorens aktive effekt $P_e = 1.0\;\text{pu}$, generatorspenning ved terminalene $V_t = 1.0\;\text{pu}$, spenning på den uendelige bussen $V_\infty = 1.0\;\text{pu}$ og generatorens transiente reaktans $X_d' = 0.20\;\text{pu}$. Alle reaktanser er oppgitt på samme systembase.

### Fremgangsmåte

Løsningen følger disse trinnene: (i) reduser transmisjonsnettet til en ekvivalent reaktans $X_{nett}$, (ii) bestem referansevinkel, $\alpha$ mellom terminalspenning $V_t$ og svingbus $V_{\infty}$, (iii) bestem strøm, I, ut av generatoren (iv) beregn generatorens transiente indre spenning $E'$, (iv) bestem den ekvivalente overføringsreaktansen $X_{transfer}$, og (iv) bruk den klassiske effekt-vinkel-relasjonen

$$
P_e=\frac{|E'||V_\infty|}{X_{eq}}\sin\delta.
$$

![generator](../images/LectureNotePdelta_curve.jpg)

### Betydning

Dette eksemplet illustrerer standardprosedyren som benyttes i transientstabilitetsanalyser: (1) konstruere det ekvivalente nettet, (2) beregne generatorens interne spenning $E'$, (3) bestemme overføringsreaktansen $X_{transfer}$, og (4) utlede den tilhørende effekt-vinkel-kurven. Den resulterende effekt-vinkel-ligningen kombineres deretter med svingligningen (swing equation) for å vurdere rotorvinkelstabiliteten etter forstyrrelser.

# Driftstilstand

Systemet undersøkes og finner tilstandene før følgende operasjonell drift:
1. Friskt system (ingen feil)

2. En av linjene er kortsluttet (3-fasefeil) på midten

3. En av linjene er frakoblet

### Generelt resultat

For et tapsfritt nett kan den elektriske effekten skrives som

$$
P_e = P_{\max}\sin\delta,
$$

der

$$
P_{\max}=\frac{|E'||V_\infty|}{X_{transfer}}.
$$

Resten av eksemplet er viet til å bestemme $E'$ og $X_{transfer}$ fra nettdataene, og deretter sette disse verdiene inn i effekt-vinkel-ligningen.



# Fullstendig løsning

## Gitte data

$$
P_e = 1.0\;\text{pu}, \qquad |V_t| = 1.0\;\text{pu}, \qquad |V_\infty| = 1.0\;\text{pu}
$$

$$
X_d' = 0.20\;\text{pu}, \qquad X_T = 0.10\;\text{pu}
$$

$$
X_{line1}=X_{line2}=0.40\;\text{pu}
$$

## 1. Ekvivalent nettreaktans

To parallelle linjer gir

$$
X_{line}=\frac{0.40}{2}=0.20\;\text{pu}
$$

slik at

$$
X_{nett}=X_T+X_{line}=0.10+0.20=0.30\;\text{pu}.
$$

## 2. Terminalspenningens vinkel

Fra effektligningen

$$
P_e=\frac{|V_t||V_\infty|}{X_{nett}}\sin\alpha
$$

fås

$$
1.0=\frac{1.0\cdot1.0}{0.30}\sin\alpha
$$

og dermed

$$
\sin\alpha=0.30,
\qquad
\alpha=\sin^{-1}(0.30)=17.458^\circ.
$$

Terminalspenningen blir

$$
V_t=1.0\angle17.458^\circ
=0.954+j0.300.
$$

## 3. Utledning av effekt-vinkel-ligningen

Generatorstrømmen er

$$
I
=
\frac{1.0\angle 17.458^\circ - 1.0\angle 0^\circ}{j0.3}.
$$

Spenningen ved generatorterminalene på rektangulær form er

$$
1.0\angle 17.458^\circ
=
0.954+j0.300.
$$

Dermed blir

$$
I
=
\frac{(0.954+j0.300)-1.0}{j0.3}
=
\frac{-0.046+j0.300}{j0.3}
=
1.0+j0.1535.
$$

På polar form:

$$
I=1.012\angle 8.729^\circ\;\text{pu}.
$$

### Transient indre spenning

Den transiente indre spenningen beregnes som

$$
E'
=
V_t+jX_d'I.
$$

Setter inn verdiene:

$$
E'
=
(0.954+j0.300)+j0.2(1.0+j0.1535).
$$

Videre fås

$$
j0.2(1.0+j0.1535)
=
j0.2-0.0307,
$$

slik at

$$
E'
=
(0.954+j0.300)+(-0.0307+j0.2)
=
0.923+j0.500.
$$

På polar form:

$$
E'
=
1.050\angle 28.44^\circ\;\text{pu}.
$$

# Driftstilstand 1: Friskt system

### Total overføringsreaktans

Mellom den transiente indre spenningen og den uendelige bussen er den totale reaktansen

$$
X_{transfer}
=
0.2+0.1+\frac{0.4}{2}
=
0.5\;\text{pu}.
$$



### Effekt-vinkel-ligning

For et tapsfritt system gjelder

$$
P_e
=
\frac{|E'||V_\infty|}{X_{transfer}}
\sin\delta.
$$

Siden

$$
|E'|=1.050,
\qquad
|V_\infty|=1.0,
\qquad
X_{transfer}=0.5,
$$

får vi

$$
P_e
=
\frac{(1.050)(1.0)}{0.5}
\sin\delta
=
2.10\sin\delta.
$$

Dermed blir effekt-vinkel-ligningen

$$
\boxed{
P_e = 2.10\sin\delta
}
\qquad \text{pu}.
$$

der $\delta$ er rotorvinkelen målt i forhold til den uendelige bussen.


## Svingligningen og driftspunktet

Effekt-vinkel-kurven er gitt ved

$$
P_e = 2.10\sin\delta.
$$

Den mekaniske tilførte effekten er

$$
P_m = 1.0\;\text{pu}.
$$

Driftspunktet finnes der den mekaniske effekten skjærer effekt-vinkel-kurven:

$$
P_m=P_e.
$$

Dermed får vi

$$
1.0=2.10\sin\delta_0
$$

som gir

$$
\delta_0
=
\sin^{-1}\!\left(\frac{1.0}{2.10}\right)
=
28.44^\circ.
$$

Dette er generatorrotorens opprinnelige vinkelposisjon før en forstyrrelse inntreffer.

### Svingligningen

Generatorens dynamikk beskrives av

$$
\frac{H}{180f}\frac{d^2\delta}{dt^2}
=
P_m-P_e.
$$

Ved innsetting av

$$
P_m=1.0
$$

og

$$
P_e=2.10\sin\delta
$$

får vi

$$
\frac{H}{180f}\frac{d^2\delta}{dt^2}
=
1.0-2.10\sin\delta.
$$

der

- $H$ er treghetskonstanten (MJ/MVA),
- $f$ er systemfrekvensen,
- $\delta$ er rotorvinkelen i elektriske grader.

### Likevektspunkt før feil

Ved driftspunktet $\delta_0=28.44^\circ$ er

$$
P_e
=
2.10\sin(28.44^\circ)
=
1.0\;\text{pu}.
$$

Siden

$$
P_m=P_e,
$$

blir den akselererende effekten

$$
P_a=P_m-P_e=0.
$$

Dermed fås

$$
\frac{d^2\delta}{dt^2}=0.
$$

Rotoren har derfor ingen vinkelakselerasjon, og maskinen befinner seg i stasjonær likevekt før feilen oppstår.

# Driftstilstand 2: En av linjene er kortsluttet (3-fasefeil) på midten

Systemet opererer under de gitte driftsforholdene når en trefasekortslutning oppstår på den ene av linjene i prallell.

Bestem:

1. Effekt-vinkel-ligningen under feilen.
2. Den tilhørende svingligningen.

Gitt:

$$
H = 5\;\text{MJ/MVA}
$$

## Løsning


Den transiente indre spenningen til generatoren er funnet:

$$
E' = 1.05\angle 28.44^\circ,
$$


Det som gjenstår er å bestemme den ekvivalente overføringsadmittansen mellom spenningskildene.

Busser nummereres som vist i figuren under og bussadmittansmatrisen kan da skrives direkte som

$$
Y_{\text{bus}}
=
j
\begin{bmatrix}
-3.333 & 0 & 3.333 \\
0 & -7.50 & 2.50 \\
3.333 & 2.50 & -10.833
\end{bmatrix}.
$$

![deltapa](../images/LectureNoteSMIB.jpg)

## Kron-reduksjon (nodeeliminasjon)

For å eliminere node $n$ fra admittansmatrisen benyttes Kron-reduksjon:

- $n$ = node som elimineres
- $k$ = radindeks
- $i$ = kolonneindeks

Den generelle formelen er

$$
Y_{ki,\text{ny}}
=
Y_{ki,\text{gammel}}
-
\frac{Y_{kn,\text{gammel}}Y_{ni,\text{gammel}}}
{Y_{nn,\text{gammel}}}.
$$

Når node 3 elimineres beregnes eksempelvis elementet $Y_{11}$ som

$$
Y_{11,\text{ny}}
=
Y_{11,\text{gammel}}
-
\frac{Y_{13,\text{gammel}}Y_{31,\text{gammel}}}
{Y_{33,\text{gammel}}}.
$$

Setter vi inn verdiene fra admittansmatrisen

$$
Y_{11}
=
-j3.333,
\qquad
Y_{13}
=
j3.333,
\qquad
Y_{31}
=
j3.333,
\qquad
Y_{33}
=
-j10.833,
$$

får vi

$$
Y_{11,\text{ny}}
=
-j3.333
-
\frac{(j3.333)(j3.333)}
{-j10.833} = -j2.307
$$

Tilsvarende beregnes de øvrige elementene i den reduserte admittansmatrisen. Resultatet etter eliminering av node 3 blir en $2\times2$-matrise som kun beskriver koblingen mellom generatorens interne spenning $E'$ og den uendelige bussen.

![deltapa](../images/LectureNoteDeltaPa.jpg)

$$
Y_{\text{bus}}
=
j
\begin{bmatrix}
-2.308 & 0.769 \\
0.769 & -6.923
\end{bmatrix}
$$

Admittansmatrisen viser at størrelsen på overføringsadmittansen er

$$
|Y_{12}| = 0.769,
$$

og dermed blir maksimal overført effekt

$$
P = |E_1||E_2||Y_{12}|
  = (1.05)\cdot(1.0)\cdot(0.769).
$$

Feiltilfellets kraftvinkelligning blir dermed

$$
P_e = 0.808\sin\delta
$$

og svingligningen kan skrives som

$$
\frac{5}{180f}\frac{d^2\delta}{dt^2}
=
1.0 - 0.808\sin\delta.
$$

Merk at rotorens posisjon ikke kan endres momentant når feilen oppstår på grunn av treghetsmomentet. Rotorvinkelen er derfor ved feiløyeblikket fortsatt lik den opprinnelige vinkelen,

$$
\delta = \delta_0.
$$

Den elektriske effekten $P_e=0.808sin28.44^{\circ}=0.385$. Det vil si at initiell akselerasjonseffekt er 

$$P_a = 1 - 0.385 = 0.615 pu$$

#### Akselerasjonseffekt i svingeligningen

$$
    \frac{d^2\delta}{dt^2} = \frac{180\cdot f}{5} = 22.14 \text{f elektriske grader/s^2}
$$

## Driftstilstand 3: En av linjene er frakoblet (klarert feil)

Feilen fra driftstilstand 2 er klarert ved åpning effektbryter. Bestem krafteffektvinkel og svingeligning.

### Effekt-vinkel-ligning

Besten admittansen for det nye overføringen:

$$\frac{1}{X_{transfer}}=\frac{1}{j(0.2+0.1+0.4)}=-j1.429 pu$$



Den nye overføringsadmittansen

$$
|Y_{12}| = j1.429,
$$

og dermed blir maksimal overført effekt

$$
P = |E_1||E_2||Y_{12}|\cdot sin\delta
  = (1.05)\cdot(1.0)\cdot(1.429)\cdot sin\delta = 1.500 sin\delta
$$

# Plot effektvinkel kurvene ($P-\delta$-kurvene)

Under er de tre effektvinkel kurvene plottet i python. Legg merke til at initiell driftsvinkel før feil ($\delta_0=28.44^{\circ}$) treffer $P_m$ ved 1 pu. 

```python
import numpy as np
import matplotlib.pyplot as plt

Pmax_prefault = 2.10
Pmax_fault = 0.808
Pmax_openline = 1.500
Pm = 1.0

delta_deg = np.linspace(0, 180, 1000)
delta_rad = np.radians(delta_deg)
delta_0 = 28.44

Pe_prefault = Pmax_prefault * np.sin(delta_rad)
Pe_fault = Pmax_fault * np.sin(delta_rad)
Pe_openline = Pmax_openline * np.sin(delta_rad)

plt.figure(figsize=(9, 5))
plt.plot(delta_deg, Pe_prefault, 'b', lw=2,
         label=r'$P_{e_{prefault}}=2.10\sin\delta$')
plt.plot(delta_deg, Pe_openline, 'k', lw=2,
         label=r'$P_{e_{openline}}=1.500\sin\delta$')

plt.plot(delta_deg, Pe_fault, 'r', lw=2,
         label=r'$P_{e_{fault}}=0.808\sin\delta$')
plt.axhline(Pm, color='k', ls='--',
            label=r'$P_m=1.0$')
ax = plt.gca()
ax.vlines(delta_0, 0, 1,
          color='k', linestyle='--',
          label=r'$\delta_0=28.44$')

plt.xlabel(r'$\delta$ (grader)')
plt.ylabel('Effekt (p.u.)')
plt.title('Effektvinkel-kurve')
plt.grid(True)
plt.legend()
plt.show()
```
